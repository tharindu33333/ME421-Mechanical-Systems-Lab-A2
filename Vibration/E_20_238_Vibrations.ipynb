### Governing Equation for Transverse Beam Vibration

Based on the Bernoulli–Euler beam theory, the transverse displacement $v(x,t)$ of a thin beam is governed by the following partial differential equation:

$$\rho A \frac{\partial^2 v(x,t)}{\partial t^2} + c \frac{\partial v(x,t)}{\partial t} + P \frac{\partial^2 v(x,t)}{\partial x^2} + \frac{\partial^2}{\partial x^2} \left( EI \frac{\partial^2 v(x,t)}{\partial x^2} \right) = f(x,t)$$


**Variable Definitions:**

* **$\rho$**: Mass density ($kg/m^3$)
* **$A$**: Cross-sectional area ($m^2$)
* **$E$**: Young’s modulus ($N/m^2$)
* **$I$**: Second moment of area ($m^4$)
* **$c$**: Viscous damping coefficient per unit length ($N \cdot s/m^2$)
* **$P$**: Axial force (N)
* **$v(x,t)$**: Transverse displacement (m)
* **$f(x,t)$**: Distributed transverse force per unit length ($N/m$)

### Governing Equation: Localized Damping and Excitation

For a beam with a point damper (dashpot) at $l_d$ and a harmonic point exciter at $l_e$, the equation of motion is:

$$\rho A \frac{\partial^2 v(x,t)}{\partial t^2} + EI \frac{\partial^4 v(x,t)}{\partial x^4} + c \, \delta(x-l_d) \frac{\partial v(x,t)}{\partial t} = F_e \sin(\omega_e t) \, \delta(x-l_e)$$



**Parameters:**

* **$\rho$**: Mass density ($kg/m^3$)
* **$A$**: Cross-sectional area ($b \cdot h$) ($m^2$)
* **$EI$**: Flexural rigidity ($N \cdot m^2$)
* **$v(x,t)$**: Transverse displacement ($m$)
* **$c$**: Viscous damping coefficient of dashpot ($N \cdot s/m$)
* **$l_d$**: Dashpot position from reference end ($m$)
* **$F_e$**: Amplitude of harmonic point force ($N$)
* **$\omega_e$**: Forcing angular frequency ($rad/s$)
* **$l_e$**: Exciter position from reference end ($m$)
* **$\delta(x-l)$**: Dirac delta function

## Modal Analysis of Beam Vibration

### 1. Separation of Variables
Assuming the beam vibrates in a fixed shape ratio for each mode, we separate the displacement $v(x,t)$ into a spatial mode shape $Y(x)$ and a generalized time coordinate $q(t)$:

$$v(x,t) = Y(x)q(t)$$

**Temporal and Spatial Derivatives:**
$$\frac{\partial v}{\partial t} = Y(x)\dot{q}(t), \quad \frac{\partial^2 v}{\partial t^2} = Y(x)\ddot{q}(t), \quad \frac{\partial^4 v}{\partial x^4} = Y''''(x)q(t)$$

---

### 2. Governing Equations
Substituting these into the free vibration equation ($\rho A \frac{\partial^2 v}{\partial t^2} + EI \frac{\partial^4 v}{\partial x^4} = 0$) and dividing by $\rho A Y(x) q(t)$:

$$\frac{EI}{\rho A} \frac{Y''''(x)}{Y(x)} = -\frac{\ddot{q}(t)}{q(t)} = \omega^2$$

This yields two Ordinary Differential Equations (ODEs):

1. **Time Equation:** $\ddot{q}(t) + \omega^2 q(t) = 0$
2. **Spatial Equation:** $Y''''(x) - \lambda^4 Y(x) = 0$

Where the eigenvalue $\lambda$ is defined as:
$$\lambda^4 = \frac{\rho A \omega^2}{EI} \implies \omega = \lambda^2 \sqrt{\frac{EI}{\rho A}}$$

---

### 3. General Solution and Boundary Conditions
The general solution for the spatial displacement is:
$$Y(x) = C_1 \cosh(\lambda x) + C_2 \sinh(\lambda x) + C_3 \cos(\lambda x) + C_4 \sin(\lambda x)$$



#### Boundary Conditions at $x = 0$ (Pivot/Pinned):
$$Y(0) = 0, \quad Y''(0) = 0 \implies C_1 = C_3 = 0$$
Reduced form: $Y(x) = C_2 \sinh(\lambda x) + C_4 \sin(\lambda x)$

#### Boundary Conditions at $x = L$ (Spring Support):
At the end with a spring constant $k$:
1. **Moment:** $Y''(L) = 0 \implies C_2 \sinh(\lambda L) - C_4 \sin(\lambda L) = 0$
2. **Shear:** $EI Y'''(L) - k Y(L) = 0$

Solving the characteristic equation for $\lambda$ gives the frequency equation:
$$EI \lambda^3 (\coth(\lambda L) - \cot(\lambda L)) = 2k$$

---

### 4. Normalized Mode Shapes
Normalizing the amplitude with $C_4 = 1$, the $i^{th}$ mode shape $Y_i(x)$ and corresponding natural frequency $\omega_i$ are:

$$Y_i(x) = \sin(\lambda_i x) + \frac{\sin(\lambda_i L)}{\sinh(\lambda_i L)} \sinh(\lambda_i x)$$

$$\omega_i = \lambda_i^2 \sqrt{\frac{EI}{\rho A}}$$

### III. Boundary Conditions and Modal Solution

**General Solution:**
$$Y(x) = C_1 \cosh(\lambda x) + C_2 \sinh(\lambda x) + C_3 \cos(\lambda x) + C_4 \sin(\lambda x) \quad (10)$$

**At $x = 0$ (pivot):**
$$Y(0)=0, \quad Y''(0)=0 \Rightarrow C_1=C_3=0$$

**Thus:**
$$Y(x) = C_2 \sinh(\lambda x) + C_4 \sin(\lambda x)$$
$$Y'(x) = \lambda(C_2 \cosh(\lambda x) + C_4 \cos(\lambda x))$$
$$Y''(x) = \lambda^2(C_2 \sinh(\lambda x) - C_4 \sin(\lambda x))$$
$$Y'''(x) = \lambda^3(C_2 \cosh(\lambda x) - C_4 \cos(\lambda x))$$

**At $x = L$ (spring end):**
$$Y''(L)=0 \Rightarrow C_2 \sinh(\lambda L) - C_4 \sin(\lambda L) = 0$$
$$EI Y'''(L) - k Y(L) = 0 \Rightarrow EI \lambda^3 (\coth(\lambda L) - \cot(\lambda L)) = 2k$$

Solving gives $\lambda$, hence natural frequencies. Normalize with $C_4 = 1$:
$$Y_i(x) = \sin(\lambda_i x) + \frac{\sin(\lambda_i L)}{\sinh(\lambda_i L)} \sinh(\lambda_i x) \quad (11)$$
$$\omega_i = \lambda_i^2 \sqrt{\frac{EI}{\rho A}} \quad (12)$$

---

### IV. Forced Response

**Governing equation:**
$$\rho A \frac{\partial^2 v}{\partial t^2} + EI \frac{\partial^4 v}{\partial x^4} + c_d \delta(x - l_d) \frac{\partial v}{\partial t} = F_e \sin(\omega_e t) \delta(x - l_e) \quad (1)$$

**Modal expansion:**
$$v(x,t) = \sum_{i=1}^{N} Y_i(x) q_i(t) \quad (3)$$

**Substitution:**
$$\rho A \sum Y_i \ddot{q}_i + EI \sum Y_i'''' q_i + c_d \delta(x - l_d) \sum Y_i \dot{q}_i = F_0 \sin(\Omega t) \delta(x - l_e) \quad (13)$$

**Multiplying by $Y_j(x)$ and integrating:**
$$\sum M_{ji} \ddot{q}_i + \sum C_{ji} \dot{q}_i + \sum K_{ji} q_i = Q_j(t) \quad (14)$$

**Where:**
* $M_{ji} = \int_0^L \rho A Y_j Y_i \, dx$
* $C_{ji} = c_d Y_j(l_d) Y_i(l_d)$
* $K_{ji} = \int_0^L EI Y_j Y_i'''' \, dx$
* $Q_j(t) = F_0 Y_j(l_e) \sin(\Omega t)$

---

### V. Adding Lumped Masses

**Using orthogonality:**
$$\int_0^L \rho A Y_j Y_i \, dx = 0 \quad (i \neq j)$$
$$K_{ji} = \omega_i^2 M_{ji}$$

**Assuming light damping:**
$$C_{ji} \approx 0 \quad (i \neq j), \quad C_{jj} = c_d Y_j^2(l_d)$$

**Single-mode equation:**
$$M_j \ddot{q}_j + C_j \dot{q}_j + K_j q_j = Q_j(t) \quad (15)$$

**Where:**
* $M_j = \int_0^L \rho A Y_j^2 dx + m_e Y_j^2(l_e) + m_d Y_j^2(l_d) + (m_f + \alpha m_s) Y_j^2(L)$
* $C_j = c_d Y_j^2(l_d)$
* $K_j = \omega_j^2 M_j$
* $Q_j(t) = F_0 Y_j(l_e) \sin(\Omega t)$

### VI. Initial Conditions

Assuming the beam starts from rest:
$$v(x,0) = v_0(x) = 0, \quad \dot{v}(x,0) = \dot{v}_0(x) = 0$$

The modal initial conditions are derived from the orthogonality property:
$$q_j(0) = \frac{\int_0^L \rho A Y_j(x) v_0(x) \, dx + \sum_r m_r Y_j(x_r) v_0(x_r)}{M_j} = 0$$
$$\dot{q}_j(0) = \frac{\int_0^L \rho A Y_j(x) \dot{v}_0(x) \, dx + \sum_r m_r Y_j(x_r) \dot{v}_0(x_r)}{M_j} = 0$$

**Equation of Motion in Standard Form:**
$$\ddot{q}_j(t) + 2\zeta_j \omega_j \dot{q}_j(t) + \omega_j^2 q_j(t) = \frac{Q_j(t)}{M_j} \quad (16)$$

Where the damping ratio is: $\zeta_j = \frac{C_j}{2\omega_j M_j}$

#### General Time Response
The total response is the sum of homogeneous (transient) and particular (steady-state) solutions:
$$q_j(t) = q_{j,h}(t) + q_{j,p}(t) \quad (17)$$

For the underdamped case ($\zeta_j < 1$):
$$q_{j,h}(t) = e^{-\zeta_j \omega_j t} \left( A_j \cos(\omega_{dj} t) + B_j \sin(\omega_{dj} t) \right) \quad (18)$$
$$\omega_{dj} = \omega_j \sqrt{1 - \zeta_j^2}$$

Steady-state response:
$$q_{j,p}(t) = q^j \sin(\Omega t - \phi_j) \quad (19)$$

With amplitude and phase:
$$q^j = \frac{F_0 Y_j(l_e)}{M_j \sqrt{(\omega_j^2 - \Omega^2)^2 + (2\zeta_j \omega_j \Omega)^2}}$$
$$\phi_j = \tan^{-1} \left( \frac{2\zeta_j \omega_j \Omega}{\omega_j^2 - \Omega^2} \right)$$

Applying initial conditions to solve for coefficients:
$$A_j = q^j \sin(\phi_j)$$
$$B_j = \frac{\zeta_j \omega_j A_j - q^j \Omega \cos(\phi_j)}{\omega_{dj}}$$

---

### VII. General Solution

The total transverse displacement $v(x,t)$ is the summation of all participating modes:

$$v(x,t) = \sum_{i=1}^{N} \left[ \sin(\lambda_i x) + \frac{\sin(\lambda_i L)}{\sinh(\lambda_i L)} \sinh(\lambda_i x) \right] \left( e^{-\zeta_i \omega_i t} (A_i \cos(\omega_{di} t) + B_i \sin(\omega_{di} t)) + q^i \sin(\Omega t - \phi_i) \right) \quad $$

### Numerical Solution for Eigenvalues

To determine the natural frequencies of the beam–spring system, we solve for the eigenvalues $\lambda$ using the characteristic equation:

$$EI\lambda^3\big(\coth(\lambda L)-\cot(\lambda L)\big)=2k_s$$

---

**1. Dimensionless Transformation**

Define the dimensionless parameter:
$$\beta = \lambda L$$

Multiplying by $\frac{L^3}{EI}$ yields:
$$\beta^3\big(\coth(\beta)-\cot(\beta)\big) = \alpha$$

where the stiffness ratio $\alpha$ is defined as:
$$\alpha = \frac{2k_sL^3}{EI}$$

---

**2. Root Finding Procedure**

Define the root function $F(\beta)$:
$$F(\beta) = \beta^3\big(\coth(\beta)-\cot(\beta)\big) - \alpha$$

* **Bracketing:** Since $\cot(\beta)$ is periodic with singularities at $n\pi$, roots are searched in the intervals:
$$((n-1)\pi, \, n\pi)$$
* **Algorithm:** The `brentq` algorithm (from `scipy.optimize`) is used to locate zeros of $F(\beta)$.

---

**3. Natural Frequencies**

Once $\lambda_n$ is found, the natural frequencies in Hz are calculated as:
$$f_n = \frac{\omega_n}{2\pi} = \frac{\lambda_n^2}{2\pi}\sqrt{\frac{EI}{\rho A}}$$

```python
def cot(x):
    return 1 / np.tan(x)

def coth(x):
    return 1 / np.tanh(x)


def find_betas(alpha, N=2, eps=1e-6):
    """ Solve characteristic equation for beta """
    betas = []
    for n in range(1, N+1):

        a = (n-1)*np.pi + eps
        b = n*np.pi - eps

        def F(beta):
            return beta**3 * (coth(beta) - cot(beta)) - alpha

        xs = np.linspace(a, b, 2000)
        Fs = F(xs)

        idx = np.where(np.sign(Fs[:-1]) != np.sign(Fs[1:]))[0]

        if len(idx) == 0:
            raise RuntimeError(f"No root in n={n}")

        i0 = idx[0]
        root = brentq(F, xs[i0], xs[i0+1])
        betas.append(root)

    return np.array(betas)


# --- Parameters ---

alpha = 2 * k_s * L**3 / (E * I)

betas = find_betas(alpha, N)
lambdas = betas / L

beam_factor = np.sqrt((E * I) / (rho * A))
omegas = lambdas**2 * beam_factor

freqs_hz = omegas / (2 * np.pi)

df = pd.DataFrame({
    'Mode': range(1, len(betas)+1),
    'Lambda (1/m)': lambdas,
    'Omega (rad/s)': omegas,
    'Frequency (Hz)': freqs_hz
})
```

Modal Frequency Function Definition

```python
def find_updated_frequencies(
    df: pd.DataFrame,
    L: float, rho: float, A: float, E: float, I: float,
    point_masses=None,
    k_s: float = 0.0,
    n_int: int = 4000,
) -> pd.DataFrame:
    """
    Compute updated modal mass, stiffness, and frequencies
    """
    if point_masses is None:
        point_masses = []

    lambdas = df["Lambda (1/m)"].to_numpy(dtype=float)

    x = np.linspace(0.0, L, n_int)

    M_beam = np.zeros_like(lambdas)
    M_added = np.zeros_like(lambdas)
    K_beam = np.zeros_like(lambdas)
    K_spring = np.zeros_like(lambdas)
    omega_new = np.zeros_like(lambdas)

    for i, lam in enumerate(lambdas):
        Yi = _Y(lam, x)
        Yi_pp = _Ypp(lam, x)

        M_beam[i] = rho * A * np.trapezoid(Yi**2, x)

        added = 0.0
        for m_r, x_r in point_masses:
            added += m_r * (_Y(lam, x_r)**2)
        M_added[i] = added

        K_beam[i] = E * I * np.trapezoid(Yi_pp**2, x)

        if k_s > 0:
            K_spring[i] = k_s * (_Y(lam, L)**2)

        omega_new[i] = np.sqrt((K_beam[i] + K_spring[i]) / (M_beam[i] + M_added[i]))

    out = df.copy()
    out["M_beam (kg)"] = M_beam
    out["M_added (kg)"] = M_added
    out["M_total (kg)"] = M_beam + M_added
    out["K_beam (N/m)"] = K_beam
    out["K_spring (N/m)"] = K_spring
    out["K_total (N/m)"] = K_beam + K_spring
    out["Omega_new (rad/s)"] = omega_new
    out["Frequency_new (Hz)"] = omega_new / (2*np.pi)

    return out
```

Find Modal Frequencies

```python
def _Y(lam, xx):
    C = np.sin(lam * L) / np.sinh(lam * L)
    return np.sin(lam * xx) + C * np.sinh(lam * xx)

def _Ypp(lam, xx):
    C = np.sin(lam * L) / np.sinh(lam * L)
    return -(lam**2) * np.sin(lam * xx) + C * (lam**2) * np.sinh(lam * xx)


# Point masses and positions
point_masses = [
    (m_d, l_d),
    (m_e, l_e),
    (m_f, l_s),
    (m_s/3, l_s),
]

# System frequencies
df = find_updated_frequencies(
    df=df,
    L=L, rho=rho, A=A, E=E, I=I,
    point_masses=point_masses,
    k_s=k_s,
    n_int=4000
)

cols = ['Mode', 'Lambda (1/m)', 'Omega_new (rad/s)', 'Frequency_new (Hz)']
print("Table 01 : First two modes")
df[cols]
```

```python
def plot_mode_shapes(lambdas, L, normalize=True, n_points=1200):
    """ Plot beam mode shapes (enhanced style) """

    x = np.linspace(0, L, n_points)
    plt.figure(figsize=(9,5), dpi=120)

    for i, lam in enumerate(lambdas, start=1):
        Y = _Y(lam, x)
        if normalize:
            Y = Y / np.max(np.abs(Y))
        plt.plot(x, Y, linewidth=2, label=f"Mode {i}")

    plt.axhline(0, color='black', linewidth=1, linestyle='--')
    plt.xlabel("Beam Length  x  (m)", fontsize=11)
    plt.ylabel("Normalized Mode Shape", fontsize=11)
    plt.title("Beam Mode Shapes", fontsize=13, fontweight='bold')
    plt.legend(frameon=True, fontsize=10)
    plt.grid(True, linestyle=':', linewidth=0.8)
    plt.tight_layout()
    plt.show()

plot_mode_shapes(lambdas, L)
```

# Damping estimation - Mode 2

### VIII. Practical Limitations and Damping Estimation

**1. Sensor Placement Issues**
Since the displacement sensor is located at the node of mode 2, direct measurement at $\omega_2$ is not possible. Therefore, the damping ratio is obtained using the frequency response method around the second-mode resonance.

The excitation frequency is swept around Mode 2, and amplitude vs. frequency (acceleration) is recorded. Using the **half-power bandwidth method**, the damping ratio is:

$$\zeta_2 \approx \frac{f_b - f_a}{2 f_{n2}}$$

where:
* $f_a$: frequency below resonance at half-power amplitude
* $f_b$: frequency above resonance at half-power amplitude



---

**2. Practical Constraint**
The exciter motor is limited to **17 Hz**, which is below the second natural frequency ($f_{n2} = 24.70$ Hz); hence, direct measurement at the second resonance is not possible under current hardware constraints.

* **Mode 2 Parameters:** $f_{n2} = 24.70$ Hz, $\omega_2 = 155.207$ rad/s

---

**3. Proportional Damping Assumption**
To estimate damping for higher modes, we assume **Rayleigh (proportional) damping**:

$$C = \alpha M + \beta K$$

In scenarios where two distinct modal measurements are unavailable, one coefficient is typically assumed to be zero. Given that stiffness-proportional damping often dominates in beam structures:

$$\alpha = 0$$

Under this assumption, the relationship between the damping ratios of different modes simplifies to:

$$\zeta_2 = \zeta_1 \frac{\omega_2}{\omega_1}$$

```python
import numpy as np
import matplotlib.pyplot as plt

# ================== PARAMETERS ==================
m1 = 1.85
k_beam = 950
c = 1.2
m2 = 0.25
k_abs = 250

y0 = 0.02
v0 = 0

# ================== DERIVED VALUES ==================
wn = np.sqrt(k_beam / m1)
zeta = c / (2 * np.sqrt(k_beam * m1))
wd = wn * np.sqrt(1 - zeta**2)

# ================== TIME ARRAY ==================
t = np.linspace(0, 5, 1000)

# ================== FUNCTIONS ==================

def free_vibration(t):
    return np.exp(-zeta * wn * t) * (
        y0 * np.cos(wd * t) +
        ((v0 + zeta * wn * y0) / wd) * np.sin(wd * t)
    )

def forced_vibration(t, omega, F0=1.0):
    mag = F0 / np.sqrt((k_beam - m1 * omega**2)**2 + (c * omega)**2)
    phase = np.arctan2(c * omega, (k_beam - m1 * omega**2))
    return mag * np.sin(omega * t - phase)

def amplitude_2dof(w):
    Z1 = (k_beam + k_abs - m1 * w**2 + 1j * c * w)
    Z2 = (k_abs - m2 * w**2)
    Z12 = -k_abs
    return np.abs(Z2 / (Z1 * Z2 - Z12**2))

# ================== CALCULATIONS ==================

free_resp = free_vibration(t)

speeds = [0.8 * wn, wn, 1.2 * wn]
labels = ['Sub-resonant', 'Resonant', 'Super-resonant']

w_range = np.linspace(0.1 * wn, 2.5 * wn, 1000)
amp_2dof = np.array([amplitude_2dof(w) for w in w_range])

# ================== PLOTTING ==================

plt.figure(figsize=(12, 12))

# --- Free vibration ---
plt.subplot(3, 1, 1)
plt.plot(t, free_resp, label="Free Response")
plt.title("Free Vibration Response")
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.grid()
plt.legend()

# --- Forced vibration ---
plt.subplot(3, 1, 2)
for s, l in zip(speeds, labels):
    plt.plot(t, forced_vibration(t, s), label=l)

plt.title("Forced Response (Different Speeds)")
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.grid()
plt.legend()

# --- Frequency response ---
plt.subplot(3, 1, 3)
plt.plot(w_range / wn, amp_2dof, linewidth=2)
plt.axvline(1.0, linestyle='--', label="ω/ωn = 1")
plt.yscale('log')

plt.title("2-DOF Frequency Response")
plt.xlabel("Frequency Ratio (ω/ωn)")
plt.ylabel("Amplitude")
plt.grid(True, which='both')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks, savgol_filter

# ==========================================
# 1. SIMULATED DATA (same as yours)
# ==========================================
freq_data = np.linspace(10, 20, 2000)

w = 2 * np.pi * freq_data
w1 = 2 * np.pi * 13.03
w2 = 2 * np.pi * 16.50

z1_real = 0.015
z2_real = 0.025

resp1 = 1 / np.sqrt((1 - (w/w1)**2)**2 + (2*z1_real*w/w1)**2)
resp2 = 0.5 / np.sqrt((1 - (w/w2)**2)**2 + (2*z2_real*w/w2)**2)

amp_data = resp1 + resp2 + np.random.normal(0, 0.05, len(freq_data))

# ==========================================
# 2. SMOOTH DATA (KEY FIX)
# ==========================================
amp_smooth = savgol_filter(amp_data, 101, 3)

# ==========================================
# 3. HALF-POWER METHOD
# ==========================================
def calculate_damping(freqs, amps):
    print(f"{'Mode':<6} | {'fn (Hz)':<10} | {'Zeta (ζ)':<10}")
    print("-" * 35)

    # Detect only dominant peaks
    peaks, _ = find_peaks(amps, prominence=np.max(amps)*0.2, distance=200)

    results = []

    for i, p in enumerate(peaks):
        fn = freqs[p]
        Amax = amps[p]
        Ahalf = Amax / np.sqrt(2)

        # Left side
        left_idx = np.where(amps[:p] <= Ahalf)[0]
        f1 = freqs[left_idx[-1]] if len(left_idx) > 0 else freqs[0]

        # Right side
        right_idx = np.where(amps[p:] <= Ahalf)[0]
        f2 = freqs[p + right_idx[0]] if len(right_idx) > 0 else freqs[-1]

        zeta = (f2 - f1) / (2 * fn)

        results.append((fn, zeta, f1, f2, Ahalf))
        print(f"{i+1:<6} | {fn:<10.3f} | {zeta:<10.5f}")

    return results

# ==========================================
# 4. RUN
# ==========================================
print("Filtered Results:\n")
modes = calculate_damping(freq_data, amp_smooth)

# ==========================================
# 5. PLOT
# ==========================================
plt.figure(figsize=(10,6))

plt.plot(freq_data, amp_data, alpha=0.3, label='Raw Data')
plt.plot(freq_data, amp_smooth, 'b', label='Smoothed Data')

for i, (fn, zeta, f1, f2, Ahalf) in enumerate(modes):
    plt.hlines(Ahalf, f1, f2, colors='r')
    plt.plot(fn, Ahalf*np.sqrt(2), 'ro')
    plt.text(fn, Ahalf, f" ζ={zeta:.4f}")

plt.title("Improved Half-Power Bandwidth Method")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(True)

plt.show()
```

# Activity 5

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. SYSTEM DEFINITION
# ==========================================
def define_primary_system():
    m_beam = 1.65
    L_beam = 0.8
    l_s = 0.8
    l_e = 0.4
    k_s = 3800.0
    m_motor = 1.0

    K_theta = k_s * (l_s**2)
    I_A = (1/3)*m_beam*(L_beam**2) + m_motor*(l_e**2)

    omega_n = np.sqrt(K_theta / I_A)
    f_n = omega_n / (2*np.pi)

    return K_theta, I_A, omega_n, f_n, l_s, l_e

# ==========================================
# 2. ABSORBER DESIGN
# ==========================================
def design_absorber(omega_n, m_abs=0.2, l_a=0.75):
    k_abs = m_abs * omega_n**2
    return m_abs, k_abs, l_a

# ==========================================
# 3. FREQUENCY RESPONSE
# ==========================================
def compute_response(freqs, K_theta, I_A, l_s, l_e, m_abs, k_abs, l_a):

    response_no_abs = []
    response_with_abs = []

    # 2-DOF matrices
    M = np.array([[I_A, 0],
                  [0, m_abs]])

    K = np.array([[K_theta + k_abs*(l_a**2), -k_abs*l_a],
                  [-k_abs*l_a,               k_abs]])

    C = np.array([[2.0, 0],
                  [0, 0.5]])

    m_unbal = 0.009

    for f in freqs:
        w = 2*np.pi*f
        F = m_unbal * w**2

        # -------- 1-DOF --------
        Z1 = K_theta - w**2*I_A + 1j*w*2.0
        theta = (F * l_e) / Z1
        response_no_abs.append(abs(theta) * l_s * 1000)

        # -------- 2-DOF --------
        Z2 = K - w**2*M + 1j*w*C
        F_vec = np.array([F * l_e, 0])

        Q = np.linalg.solve(Z2, F_vec)
        response_with_abs.append(abs(Q[0]) * l_s * 1000)

    return np.array(response_no_abs), np.array(response_with_abs)

# ==========================================
# 4. MAIN EXECUTION
# ==========================================
K_theta, I_A, omega_n, f_target, l_s, l_e = define_primary_system()

print(f"Target Resonance Frequency: {f_target:.2f} Hz")

m_abs, k_abs, l_a = design_absorber(omega_n)

print(f"Absorber stiffness required: {k_abs:.2f} N/m")

freqs = np.linspace(5, 15, 500)

resp_no, resp_with = compute_response(
    freqs, K_theta, I_A, l_s, l_e, m_abs, k_abs, l_a
)

# ==========================================
# 5. PLOTTING
# ==========================================
plt.figure(figsize=(10,6))

plt.plot(freqs, resp_no, '--', linewidth=2, label='Without Absorber')
plt.plot(freqs, resp_with, '-', linewidth=2, label='With Tuned Absorber')

plt.axvline(f_target, linestyle=':', label='Tuned Frequency')

plt.xlabel("Frequency (Hz)")
plt.ylabel("Tip Amplitude (mm)")
plt.title("Vibration Absorber Performance")

plt.grid(True)
plt.legend()

# Highlight notch
idx = np.argmin(np.abs(freqs - f_target))
plt.scatter(freqs[idx], resp_with[idx])

plt.annotate("Anti-resonance",
             xy=(freqs[idx], resp_with[idx]),
             xytext=(freqs[idx], resp_with[idx] + 5),
             arrowprops=dict())

plt.show()
```
