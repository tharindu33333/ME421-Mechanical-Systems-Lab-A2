### derive the 2-dof freedom model that will capture the first two dominant modes of small amplitude vibration of the system.

The TM1016V simply supported beam with vibration absorber forms a 2-DOF system for small-amplitude vibrations, capturing the first two dominant modes.

This equation describes small vibrations in a slender beam.


$
\rho A{\partial^2 y \over {\partial t^2}} + c{\partial y \over {\partial t}}+EI{{\partial^4 y} \over {\partial x^4}}+P {\partial^2 y \over {\partial x^2}}=f(t,x)
$


where:

$P$ is the axial compressive force,

$\rho$ is the material density,

$E$ is the elastic modulus,

$I$ is the moment of inertia,

$A$ is the cross-sectional area,

$c$ is the distributed viscous damping, and

$f(x,t)$ is the external transverse load.


#### Since  $P$ = 0  for this case

$
\rho A{\partial^2 y \over {\partial t^2}} + c{\partial y \over {\partial t}}+EI{{\partial^4 y} \over {\partial x^4}}=f(t,x)
$

boundry conditions are


$y$(0,$t$) = 0,

${\partial^3 y \over \partial x^3}(l_{beam},t)$ = 0,

$EI{\partial^2 y \over \partial x^2}(0,t)-k_{\theta}{\partial y \over \partial x}(0,t)$ = 0,    where $k_{\theta} = k_{spring}l^2_{spring}$ is the rotational stiffness

${\partial^2 y \over \partial x^2}(l_{beam},t)$ = 0,

We define a linear operator $H$ on the vector space $\mathfrak{F}$ (comprising functions that are twice differentiable over the interval $[0, L]$) to represent the structural stiffness normalized by mass:

$$ H = \frac{EI}{\rho A} \frac{\partial^4}{\partial x^4} $$

Using this operator, the governing equation of motion can be expressed in the compact form:

$$ \frac{\partial^2 y}{\partial t^2} + C \frac{\partial y}{\partial t} + H y = q(t,x) $$

where the distributed damping is

$$c = C \rho A$$

and the external force density is

$$f(t) = \rho A q(t)$$

The operator $H$ maps the transverse deflection profile to the acceleration induced by bending moments, satisfying the system's boundary conditions. We assume $H$ possesses a set of distinct, positive eigenvalues denoted by $\{\omega_k^2\}_{k=1}^\infty$ and corresponding eigenvectors (mode shapes) denoted by $\{\psi_k\}_{k=1}^\infty$.

The resulting eigenvalue problem is written as:

$$ H \psi_k(x) = \omega_k^2 \psi_k(x) $$

Substituting the definition of $H$ into this equation yields the differential equation for the mode shapes:

\begin{align*}
\frac{EI}{\rho A} \frac{d^4 \psi_k(x)}{d x^4} &= \omega_k^2 \psi_k(x) \\
\frac{d^4 \psi_k(x)}{d x^4} &= \left( \frac{\rho A \omega_k^2}{EI} \right) \psi_k(x) \\
\frac{d^4 \psi_k(x)}{d x^4} &= \beta_k^4 \psi_k(x)
\end{align*}

where the spectral parameter is defined as

$$\beta_k^4 = \frac{\rho A \omega_k^2}{EI}$$

The general solution to the fourth-order governing differential equation (1) is expressed as a linear combination of trigonometric and hyperbolic functions:

\begin{align*}
\psi_k(x) = A_k \cosh(\beta_k x) + B_k \sinh(\beta_k x) + C_k \cos(\beta_k x) + D_k \sin(\beta_k x)
\end{align*}

where the coefficients $A_k$, $B_k$, $C_k$, and $D_k$ are determined by the system's boundary conditions.

\textbf{Boundary Conditions at the Base ($x=0$):}

The displacement condition at the pivot, $\psi_k(0) = 0$, implies that $A_k + C_k = 0$, or simply $C_k = -A_k$.

The rotational constraint involves the balance of bending moment and spring torque. By evaluating the first and second derivatives at $x=0$:

\begin{align*}
\psi_k'(0) &= \beta_k(B_k+D_k) \\
\psi_k''(0) &= \beta_k^2(A_k-C_k) = 2\beta_k^2 A_k \quad (\text{since } C_k = -A_k)
\end{align*}

Substituting these into the moment boundary condition equation yields:

\begin{align*}
EI (2\beta_k^2 A_k) - k_{\theta}\beta_k(B_k+D_k) &= 0 \\
2EI \beta_k A_k - k_{\theta}(B_k+D_k) &= 0
\end{align*}

\textbf{Boundary Conditions at the Free End ($x=l_{beam}$):}

At the free end, both the bending moment and shear force must vanish. Applying $C_k = -A_k$ to the second and third derivatives yields the following relations:

\begin{align*}
\text{Zero Moment: } \quad \psi_k''(l_{beam}) = 0 \quad \Rightarrow \quad & A_k[\cosh(\beta_k l_{beam}) + \cos(\beta_k l_{beam})] + B_k\sinh(\beta_k l_{beam}) - D_k\sin(\beta_k l_{beam}) = 0 \\
\text{Zero Shear: } \quad \psi_k'''(l_{beam}) = 0 \quad \Rightarrow \quad & A_k[\sinh(\beta_k l_{beam}) - \sin(\beta_k l_{beam})] + B_k\cosh(\beta_k l_{beam}) - D_k \cos(\beta_k l_{beam}) = 0
\end{align*}

\textbf{Matrix Formulation:}

The three governing equations derived above can be assembled into a homogeneous linear system:

$$
\mathbf{H}(\beta_k)\begin{bmatrix} A_k \\ B_k \\ D_k \end{bmatrix} = \mathbf{0}
$$

where $\mathbf{H}(\beta_k)$ is a $3 \times 3$ coefficient matrix.

To obtain non-trivial solutions (where the beam undergoes vibration), the determinant of this matrix must vanish: $\text{det}(\mathbf{H}(\beta_k)) = 0$. This characteristic equation is solved numerically to determine the eigenvalues $\beta_k$ for each mode $k \in \mathbb{N}$, which are then used to solve the linear system for the mode shape coefficients $A_k$, $B_k$, and $D_k$.

```python
import numpy as np
from numpy.linalg import det
from scipy.linalg import svd
from scipy.linalg import eigh
from scipy.optimize import brentq
from scipy.integrate import odeint
from scipy.integrate import quad
import matplotlib.pyplot as plt
from IPython.display import HTML
```

```python
# ========= Determinent of the Characteristic Matrix =========
def char_det(beta, l_beam, EI, k_theta):

    bL = beta * l_beam

    H = np.array([
        [2*EI*beta**2,      -k_theta*beta,  -k_theta*beta],
        [np.cosh(bL)+np.cos(bL), np.sinh(bL), -np.sin(bL)],
        [np.sinh(bL)-np.sin(bL), np.cosh(bL), -np.cos(bL)]
    ])

    return det(H)

def find_betas(l_beam, EI, k_theta, n_roots=2, beta_max=50):
    betas = []
    beta_vals = np.linspace(1e-4, beta_max, 5000)

    f_vals = [char_det(b, l_beam, EI, k_theta) for b in beta_vals]

    for i in range(len(beta_vals)-1):
        if f_vals[i]*f_vals[i+1] < 0:
            root = brentq(
                char_det,
                beta_vals[i],
                beta_vals[i+1],
                args=(l_beam, EI, k_theta)
            )
            betas.append(root)

            if len(betas) >= n_roots:
                break

    return np.array(betas)

# ============ Solving for A,B,C,D Coefficients ===========
def mode_shape_coefficients(beta, l_beam, EI, k_theta):
    bL = beta * l_beam

    H = np.array([
        [2*EI*beta**2,      -k_theta*beta,  -k_theta*beta],
        [np.cosh(bL)+np.cos(bL), np.sinh(bL), -np.sin(bL)],
        [np.sinh(bL)-np.sin(bL), np.cosh(bL), -np.cos(bL)]
    ])

    # SVD -> right singular vector corresponding to smallest singular value
    U, S, Vh = svd(H)
    v = Vh[-1, :].copy()   # v ~ [A, B, D] up to sign/scale

    A, B, D = v
    C = -A
    return A, B, C, D
```

The spectrum of the operator $H$ is defined by the squared natural frequencies

$$
[
\{\omega_k^2\}_{k=1}^{\infty}
=
\left\{
\frac{EI}{\rho A}\,\beta_k^{4}
\right\}_{k=1}^{\infty}.
]
$$

The corresponding eigenfunctions (mode shapes) associated with $H$ are expressed as


$$
[
\{\psi_k(x)\}_{k=1}^{\infty}
=
\left\{
A_k \cosh(\beta_k x)
+ B_k \sinh(\beta_k x)
+ C_k \cos(\beta_k x)
+ D_k \sin(\beta_k x)
\right\}_{k=1}^{\infty}.
]
$$


The coefficients $A_k$, $B_k$, $C_k$, and $D_k$ are evaluated numerically using the computational procedures implemented in the preceding functions.


**Forced Vibration Response using Modal Expansion:**

> Add blockquote

When the oscillator is activated, the system undergoes forced vibration due to the applied oscillatory force $Q$.
Screenshot 2026-01-29 214527.png

Assuming the absence of base excitation, the external loading acting on the beam is taken as

$$
q(x,t) = Q \sin(\Omega t)\,\delta\!\left(x - l_{\text{exciter}}\right)
$$

The dynamics of the system may be represented as a set of independent, forced and damped
single-degree-of-freedom oscillators, one for each vibration mode $k$, such that
$$
[
\ddot{y} + C\dot{y} + Hy = q(x,t)
\quad \Longrightarrow \quad
m_k \ddot{z}_k
+ 2\zeta_k \omega_k m_k \dot{z}_k
+ m_k \omega_k^2 z_k
=
\langle\!\langle \psi_k(x), q(x,t) \rangle\!\rangle
]
$$

Here, the modal damping ratio is defined as

$$
[
\zeta_k = \frac{C}{2\omega_k}
]
$$

and the modal mass is given by
$$
[
m_k = \int_{0}^{l_{\text{beam}}} \rho A \, \psi_k^2(x)\ dx.
]
$$

The introduction of modal masses removes the arbitrary scaling associated with the choice
of the mode shape coefficients $B_k$ and $D_k$.

Solving the resulting decoupled ordinary differential equation for each mode $k$ leads to
the steady-state response
$$
[
z_k(t) = \chi_k(\Omega)\cos\!\bigl(\Omega t + \phi_k(\Omega)\bigr)
]
$$

where $\chi_k(\Omega)$ denotes the steady-state amplitude and
$\phi_k(\Omega)$ represents the phase lag of the $k^{\text{th}}$ modal coordinate
at the excitation frequency $\Omega$.


We now derive explicit expressions for the quantities defined above.

The mode shapes are orthonormal under the inner product defined as
\begin{align*}
\langle\!\langle \psi_k(x), q(x,t) \rangle\!\rangle
&= \int_{0}^{l_{\text{beam}}} \psi_k(x)\, q(x,t)\, dx
\end{align*}

Substituting the harmonic point excitation yields
\begin{align*}
\langle\!\langle \psi_k(x), q(x,t) \rangle\!\rangle
&= \int_{0}^{l_{\text{beam}}}
\psi_k(x)\, Q \sin(\Omega t)\, \delta\!\left(x - l_{\text{exciter}}\right)\, dx \\
&= Q \sin(\Omega t)
\int_{0}^{l_{\text{beam}}}
\psi_k(x)\, \delta\!\left(x - l_{\text{exciter}}\right)\, dx \\
&= Q \sin(\Omega t)\, \psi_k\!\left(l_{\text{exciter}}\right).
\end{align*}

Using this result, the steady-state modal amplitude is defined as
\begin{align*}
\chi_k(\Omega)
&= \left|
\frac{\langle\!\langle \psi_k(x), q(x,t) \rangle\!\rangle}
{m_k \sqrt{(\omega_k^2 - \Omega^2)^2
+ 4\zeta_k^2 \Omega^2 \omega_k^2}}
\right| \\
&= \left|
\frac{Q\, \psi_k(l_{\text{exciter}})}
{m_k \sqrt{(\omega_k^2 - \Omega^2)^2
+ 4\zeta_k^2 \Omega^2 \omega_k^2}}
\right|
\end{align*}

The corresponding phase lag of the $k^{\text{th}}$ modal coordinate is given by
$$[
\phi_k(\Omega)
=
\arctan\!\left(
\frac{2\zeta_k \Omega \omega_k}{\omega_k^2 - \Omega^2}
\right)
]$$

The resulting response of the system, preserving its infinite-dimensional
representation, can be expressed as
\begin{align*}
y(x,t)
&= \sum_{k=1}^{\infty} z_k(t)\,\psi_k(x) \\
&= \sum_{k=1}^{\infty}
\chi_k(\Omega)\cos\!\bigl(\Omega t + \phi_k(\Omega)\bigr)\,\psi_k(x).
\end{align*}


**Reduction to 2-DOF Model with Localised Spring–Damper Attachment**

Truncating the infinite-dimensional representation to the first two dominant
modes, the displacement field may be approximated as
\begin{align*}
y(x,t) \approx\;
&\chi_1(\Omega)\cos\!\bigl(\Omega t + \phi_1(\Omega)\bigr)\,\psi_1(x) \\
&+ \chi_2(\Omega)\cos\!\bigl(\Omega t + \phi_2(\Omega)\bigr)\,\psi_2(x).
\end{align*}

The corresponding reduced-order two-degree-of-freedom (2-DOF) system can be
expressed in matrix form as
\begin{align*}
\underbrace{
\begin{bmatrix}
m_1 & 0 \\
0 & m_2
\end{bmatrix}}_{M_{\text{beam}}}
\begin{bmatrix}
\ddot{z}_1 \\
\ddot{z}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
2\zeta_1 \omega_1 m_1 & 0 \\
0 & 2\zeta_2 \omega_2 m_2
\end{bmatrix}}_{C_{\text{beam}}}
\begin{bmatrix}
\dot{z}_1 \\
\dot{z}_2
\end{bmatrix}
+
\underbrace{
\begin{bmatrix}
\omega_1^2 m_1 & 0 \\
0 & \omega_2^2 m_2
\end{bmatrix}}_{K_{\text{beam}}}
\begin{bmatrix}
z_1 \\
z_2
\end{bmatrix}
=
\underbrace{
\begin{bmatrix}
Q\,\psi_1(l_{\text{exciter}})\sin(\Omega t) \\
[1mm]
Q\,\psi_2(l_{\text{exciter}})\sin(\Omega t)
\end{bmatrix}}_{F_{\text{exciter}}}.
\end{align*}

We now proceed to incorporate the effects of the attached spring and viscous damper.

$$\textbf{Part I: Consideration of Additional Masses}$$

For improved accuracy, the formulation accounts for the portion of the spring
mass that participates in the system motion. According to Rayleigh’s theory,
one third of the spring mass contributes to the effective inertia, and the mass
of the spring–beam attachment must also be included. The effective mass is
therefore defined as
$$
[
m_{\text{eff}} = \frac{1}{3} m_{\text{spring}} + m_{\text{fixed}}.
]
$$
The transverse acceleration of the beam at the spring attachment location is
$$
[
\ddot{y}(l_{\text{spring}},t)
=
\ddot{z}_1(t)\,\psi_1(l_{\text{spring}})
+
\ddot{z}_2(t)\,\psi_2(l_{\text{spring}}).
]
$$

The inertia force exerted by the attached mass is given by
$$[
F_{\text{inertia}}(t)
=
m_{\text{eff}}\,\ddot{y}(l_{\text{spring}},t).
]
$$

The modal force acting on the $k^{\text{th}}$ mode is
$$[
F^{(m)}_k
=
F_{\text{inertia}}(t)\,\psi_k(l_{\text{spring}}).
]
$$

Projecting this force onto the modal coordinates yields
$$
[
\begin{bmatrix}
F^{(m)}_1 \\
F^{(m)}_2
\end{bmatrix}
=
M_{\text{spring}}
\begin{bmatrix}
\ddot{z}_1 \\
\ddot{z}_2
\end{bmatrix},
]
$$

where the additional mass matrix associated with the spring attachment is

$$
[
M_{\text{spring}}
=
m_{\text{eff}}
\begin{bmatrix}
\psi_1^2(l_{\text{spring}}) &
\psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) \\
\psi_1(l_{\text{spring}})\psi_2(l_{\text{spring}}) &
\psi_2^2(l_{\text{spring}})
\end{bmatrix}.
]
$$

Similarly, the mass contributions of the damper and exciter attachments are
$$
[
M_{\text{damper}}
=
m_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix},
]
$$
and
$$
[
M_{\text{exciter}}
=
m_{\text{exciter}}
\begin{bmatrix}
\psi_1^2(l_{\text{exciter}}) &
\psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) \\
\psi_1(l_{\text{exciter}})\psi_2(l_{\text{exciter}}) &
\psi_2^2(l_{\text{exciter}})
\end{bmatrix}.
]
$$

**Part II: Consideration of Damper Attachment**

By applying the constitutive relation of a viscous damper, the damping force is
given by

$$
[
F_{\text{damper}}(t)
=
c_{\text{damper}}\,\dot{y}(l_{\text{damper}},t).
]
$$

Using the two-mode approximation, the velocity at the damper attachment point
can be expressed as
$$
[
\dot{y}(l_{\text{damper}},t)
=
\dot{z}_1(t)\,\psi_1(l_{\text{damper}})
+
\dot{z}_2(t)\,\psi_2(l_{\text{damper}}).
]
$$

The modal force acting on the $k^{\text{th}}$ mode is obtained by projection as
$$
[
F^{(c)}_k(t)
=
c_{\text{damper}}
\left(
\dot{z}_1(t)\,\psi_1(l_{\text{damper}})
+
\dot{z}_2(t)\,\psi_2(l_{\text{damper}})
\right)
\psi_k(l_{\text{damper}}).
]
$$

This results in the matrix representation
$$
[
\begin{bmatrix}
F^{(c)}_1 \\
F^{(c)}_2
\end{bmatrix}
=
C_{\text{damper}}
\begin{bmatrix}
\dot{z}_1 \\
\dot{z}_2
\end{bmatrix},
]
$$
where the damping matrix associated with the viscous damper is
$$
[
C_{\text{damper}}
=
c_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix}.
]
$$

The dynamic and inertial effects of the spring, damper, and exciter can now be
incorporated into the reduced two-degree-of-freedom system.


The final reduced-order system equation can be written as
$$
[
\left(
\mathbf{M}_{\text{beam}}
+
\mathbf{M}_{\text{spring}}
+
\mathbf{M}_{\text{damper}}
+
\mathbf{M}_{\text{exciter}}
\right)\,\ddot{\mathbf{z}}
+
\left(
\mathbf{C}_{\text{beam}}
+
\mathbf{C}_{\text{damper}}
\right)\,\dot{\mathbf{z}}
+
\mathbf{K}_{\text{beam}}\,\mathbf{z}
=
\mathbf{F}_{\text{exciter}}.
]
$$

Here,
$$
[
\mathbf{M},\ \mathbf{C},\ \mathbf{K} \in \mathbb{R}^{2\times 2},
\qquad
\mathbf{z},\ \mathbf{F} \in \mathbb{R}^{2}.
]
$$

```python
# ================= Beam Properties =================
b = 0.025          # Breadth (m)
h = 0.01           # Thickness (m)
A = b*h            # Cross-sectional area (m^2)
m_beam = 1.65      # Mass of the beam (m)
l_beam = 0.815     # Beam length (m)
E = 2e+11          # Modulus of Elasticity (Pa)
I = b * h**3 /12   # Area Moment of Inertia (m^4)
rho = m_beam/(A*l_beam) # Density (kg/m^3)

# ================= Spring Properties ==============
k_spring = 3800   # Spring stiffness (N/m)
l_spring = 0.75   # Location of spring along beam (m)
m_spring = 0.388  # Spring mass (kg)
m_fixed = 0.009   # Fixed mass attached to spring (kg)

# Rotational spring at base
k_theta = k_spring * l_spring * l_spring

# ================= Damper Properties ==============
l_damper = 0.2    # Location of damper along beam (m)
m_damper = 0.53   # Mass associated with damper (kg)

# ================= Exciter Properties =============
m_exciter = 4.2   # Exciter mass (kg)
l_exciter = 0.4   # Exciter location along beam (m)
```

```python
# ================= Find first two betas =================
roots = find_betas(l_beam, E*I, k_theta, n_roots=2)
beta1, beta2 = roots
print(f"beta1 = {beta1:.4f}, beta2 = {beta2:.4f}")

# ================= Mode shape function =================
def psi(x, beta, l_beam, EI, k_theta):
    A, B, C, D = mode_shape_coefficients(beta, l_beam, EI, k_theta)
    return A*np.cosh(beta*x) + B*np.sinh(beta*x) + C*np.cos(beta*x) + D*np.sin(beta*x)

# ================= Evaluate psi at specific points =================
psi1_spring = psi(l_spring, beta1, l_beam, E*I, k_theta)
psi2_spring = psi(l_spring, beta2, l_beam, E*I, k_theta)

psi1_damper = psi(l_damper, beta1, l_beam, E*I, k_theta)
psi2_damper = psi(l_damper, beta2, l_beam, E*I, k_theta)

psi1_exciter = psi(l_exciter, beta1, l_beam, E*I, k_theta)
psi2_exciter = psi(l_exciter, beta2, l_beam, E*I, k_theta)

print(f"psi1(l_spring) = {psi1_spring:.4f}, psi2(l_spring) = {psi2_spring:.4f}")
print(f"psi1(l_damper) = {psi1_damper:.4f}, psi2(l_damper) = {psi2_damper:.4f}")
print(f"psi1(l_exciter) = {psi1_exciter:.4f}, psi2(l_exciter) = {psi2_exciter:.4f}")

# ================= Modal masses =================
def modal_mass(beta):
    """Integrate rho*A*psi^2 along the beam"""
    integrand = lambda x: rho * A * psi(x, beta, l_beam, E*I, k_theta)**2
    m, _ = quad(integrand, 0, l_beam)
    return m

m1 = modal_mass(beta1)
m2 = modal_mass(beta2)

# ================= Natural frequencies (Beam Only) =================
omega1 = (beta1**2) * np.sqrt(E*I/(rho*A))
omega2 = (beta2**2) * np.sqrt(E*I/(rho*A))

print(f"Modal mass m1 = {m1:.4f} kg, m2 = {m2:.4f} kg")
print(f"Natural frequencies for beam alone: omega1 = {omega1:.4f} rad/s, omega2 = {omega2:.4f} rad/s")
```

```python
# ========= BEAM MATRICES =========
M_beam = np.array([[m1, 0.0],
                   [0.0, m2]])
K_beam = np.array([[m1 * omega1**2, 0.0],
                   [0.0, m2 * omega2**2]])

# ========= SPRING MASS =========
m_eff_spring = (1/3) * m_spring + m_fixed
M_spring = m_eff_spring * np.array([[psi1_spring**2, psi1_spring * psi2_spring],
                                    [psi1_spring * psi2_spring, psi2_spring**2]])

# ========= DAMPER MASS =========
M_damper = m_damper * np.array([[psi1_damper**2, psi1_damper * psi2_damper],
                                [psi1_damper * psi2_damper, psi2_damper**2]])

# ========= EXCITER MASS =========
M_exciter = m_exciter * np.array([[psi1_exciter**2, psi1_exciter * psi2_exciter],
                                 [psi1_exciter * psi2_exciter, psi2_exciter**2]])

def print_matrix(name, mat):
    print(f"{name}:")
    for row in mat:
        print("  [" + ", ".join(f"{val:10.3f}" for val in row) + "]")
    print()  # blank line for spacing

print_matrix("M_beam", M_beam)
print_matrix("M_spring", M_spring)
print_matrix("M_damper", M_damper)
print_matrix("M_exciter", M_exciter)
print_matrix("K_beam", K_beam)
```

**Activity 2: Estimation of Damping Ratios of the Two Dominant Modes**

Recall the governing equation of the system with the attachments.

$$
 \mathbf{M}\mathbf{\ddot{z}} + \mathbf{C}\mathbf{\dot{z}} + \mathbf{K}\mathbf{z} = \mathbf{F_{\text{exciter}}}
$$

where $\mathbf{M} = \mathbf{M_\text{beam}} + \mathbf{M_\text{spring}} + \mathbf{M_\text{damper}} + \mathbf{M_\text{exciter}}$, $\mathbf{C} = \mathbf{C_\text{beam} + \mathbf{C_{\text{damper}}}} $ and $\mathbf{K} = \mathbf{K_{\text{beam}}}$


Consider the free-vibration response of the previously derived dynamic system,
$$
[
\mathbf{M}\,\ddot{\mathbf{z}} + \mathbf{C}\,\dot{\mathbf{z}} + \mathbf{K}\,\mathbf{z} = 0.
]
$$
Assume a harmonic form of the solution given by
$$
[
\mathbf{z}(t) = \boldsymbol{\phi}\,e^{j\omega t},
]
$$
and neglect damping in order to determine the natural frequencies and mode
shapes of the system.

Substituting this assumed solution into the governing equation leads to
$$
[
(\mathbf{K} - \omega^2 \mathbf{M})\,\boldsymbol{\phi} = 0,
]
$$
with the corresponding characteristic equation
$$
[
\det(\mathbf{K} - \omega^2 \mathbf{M}) = 0.
]
$$

Solving this eigenvalue problem yields two natural frequencies,
$\omega_1$ and $\omega_2$, together with their associated eigenvectors
$\boldsymbol{\phi}_1$ and $\boldsymbol{\phi}_2$, where
$
\boldsymbol{\phi} \in \mathbb{R}^{2}.
$

The modal matrix is defined as

$$
\Phi = \begin{bmatrix} \phi_1 \\ \phi_2 \end{bmatrix}
$$


We now proceed to compute the numerical values of the above quantities.


```python
# ================= Natural frequencies (Full System) =================
M = M_beam + M_spring + M_damper + M_exciter
K = K_beam

eigvals, eigvecs = eigh(K, M)
omega_n = np.sqrt(eigvals)

print("Natural frequencies including attachments:")
for i, w in enumerate(omega_n, 1):
    print(f"Mode {i}: {w:.3f} rad/s = {w/(2*np.pi):.3f} Hz")

Phi = eigvecs
print_matrix("Modal Matrix", Phi)
```

Introduce a new set of generalized coordinates $\mathbf{q}(t)$ such that
$$
\mathbf{z}(t) = \boldsymbol{\Phi}\,\mathbf{q}(t).
$$

Using this transformation, the forced vibration equation of the system can be
written as
$$
\mathbf{M}\boldsymbol{\Phi}\,\ddot{\mathbf{q}}
+
\mathbf{C}\boldsymbol{\Phi}\,\dot{\mathbf{q}}
+
\mathbf{K}\boldsymbol{\Phi}\,\mathbf{q}
=
\mathbf{F}_{\text{exciter}}.
$$

Premultiplying the above equation by $\boldsymbol{\Phi}^T$, the modal matrices
are defined as
$$
\underbrace{\boldsymbol{\Phi}^T \mathbf{M} \boldsymbol{\Phi}}_{\mathbf{M}_m}
\,\ddot{\mathbf{q}}
+
\underbrace{\boldsymbol{\Phi}^T \mathbf{C} \boldsymbol{\Phi}}_{\mathbf{C}_m}
\,\dot{\mathbf{q}}
+
\underbrace{\boldsymbol{\Phi}^T \mathbf{K} \boldsymbol{\Phi}}_{\mathbf{K}_m}
\,\mathbf{q}
=
\underbrace{\boldsymbol{\Phi}^T \mathbf{F}_{\text{exciter}}}_{\mathbf{F}_m}.
$$

This results in the modal equation of motion
$$
\mathbf{M}_m\,\ddot{\mathbf{q}}
+
\mathbf{C}_m\,\dot{\mathbf{q}}
+
\mathbf{K}_m\,\mathbf{q}
=
\mathbf{F}_m.
$$

For undamped eigenvectors associated with the matrix pair
$(\mathbf{K}, \mathbf{M})$, the modal mass and stiffness matrices satisfy
$$
\mathbf{M}_m \ \text{and} \ \mathbf{K}_m \ \text{are diagonal}.
$$

For the free-vibration response, the governing modal equations reduce to
$$
\mathbf{M}_m\,\ddot{\mathbf{q}}
+
\mathbf{C}_m\,\dot{\mathbf{q}}
+
\mathbf{K}_m\,\mathbf{q}
=
0.
$$

In the modal formulation, each mode behaves as an independent single-degree-of-freedom
oscillator provided that damping does not introduce modal coupling, that is,
when the modal damping matrix $\mathbf{C}_m$ is diagonal.

Using the relation
$$
\zeta_k = \frac{c}{2 m_k \omega_k},
$$
the modal damping ratio for the complete system can be expressed as
$$
\zeta_k
=
\frac{\boldsymbol{\phi}_k^{T}\,\mathbf{C}\,\boldsymbol{\phi}_k}
{2\,\omega_k\,\boldsymbol{\phi}_k^{T}\,\mathbf{M}\,\boldsymbol{\phi}_k}.
\tag{2}
$$

Since the total damping matrix
$$
\mathbf{C} = \mathbf{C}_{\text{beam}} + \mathbf{C}_{\text{damper}}
$$
is not known a priori, an appropriate damping model must be constructed.

It is already known that the beam damping matrix $\mathbf{C}_{\text{beam}}$ is
diagonal from the modal reduction of the beam alone, while the damper
contribution is given by
$$
\mathbf{C}_{\text{damper}}
=
c_{\text{damper}}
\begin{bmatrix}
\psi_1^2(l_{\text{damper}}) &
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) \\
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) &
\psi_2^2(l_{\text{damper}})
\end{bmatrix}.
$$

Consider the mode shapes of the beam alone.


```python
# ================= Modes Shapes of the Beam Only =================
x = np.linspace(0, l_beam, 1000)

psi1 = psi(x, beta1, l_beam, E*I, k_theta)
psi2 = psi(x, beta2, l_beam, E*I, k_theta)

psi1_norm = psi1 / np.max(np.abs(psi1))
psi2_norm = psi2 / np.max(np.abs(psi2))

plt.figure(figsize=(7, 3))
plt.plot(x, psi1_norm, label=r'$\psi_1$ (Mode 1)')
plt.plot(x, psi2_norm, label=r'$\psi_2$ (Mode 2)')

plt.xlabel('Position along Beam)')
plt.ylabel('Normalised Amplitude')
plt.grid(True)
plt.legend()
plt.show()
```

From the results obtained above, it can be observed that at
$$
l_{\text{damper}} = 0.2~\text{m},
$$
the magnitude of the first mode shape is much smaller than that of the second
mode shape, that is,
$$
\lvert \psi_1(l_{\text{damper}}) \rvert \ll \lvert \psi_2(l_{\text{damper}}) \rvert.
$$

Specifically, the computed values are
$$
\psi_1(l_{\text{damper}}) = -0.2491,
\qquad
\psi_2(l_{\text{damper}}) = 0.7102.
$$

Hence,
$$
\psi_2^2(l_{\text{damper}}) = 0.5044,
\qquad
\psi_1(l_{\text{damper}})\psi_2(l_{\text{damper}}) = -0.1769.
$$

Therefore, since
$$
\lvert \psi_1(l_{\text{damper}}) \rvert \ll \lvert \psi_2(l_{\text{damper}}) \rvert,
$$
the off-diagonal terms of the damping matrix
$$
\mathbf{C}_{\text{damper}}
$$
may be neglected, implying that the modal coupling introduced by the damper is
small.

Accordingly, the system damping is assumed to follow the
\textbf{Rayleigh proportional damping} model, such that
$$
\mathbf{C}
=
\alpha\,\mathbf{M}
+
\beta\,\mathbf{K}.
$$

Premultiplying and postmultiplying by the modal vector $\boldsymbol{\phi}_k$
gives
$$
\boldsymbol{\phi}_k^{T}\mathbf{C}\boldsymbol{\phi}_k
=
\alpha\,\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k
+
\beta\,\boldsymbol{\phi}_k^{T}\mathbf{K}\boldsymbol{\phi}_k.
\tag{3}
$$

Recall that the undamped eigenvalue problem satisfies
$$
(\mathbf{K} - \omega^2 \mathbf{M})\,\boldsymbol{\phi} = 0,
$$
which implies
$$
\boldsymbol{\phi}_k^{T}\mathbf{K}\boldsymbol{\phi}_k
=
\omega_k^2\,\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k.
$$

Substituting this result into equation $(3)$ yields
$$
\boldsymbol{\phi}_k^{T}\mathbf{C}\boldsymbol{\phi}_k
=
\alpha\,\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k
+
\beta\,\omega_k^2\,\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k,
$$
or equivalently,
$$
\boldsymbol{\phi}_k^{T}\mathbf{C}\boldsymbol{\phi}_k
=
(\alpha + \beta\,\omega_k^2)\,
\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k.
$$

Substituting this expression into the modal damping ratio definition in
equation $(2)$ gives
$$
\zeta_k
=
\frac{
(\alpha + \beta\,\omega_k^2)\,
\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k
}{
2\,\omega_k\,
\boldsymbol{\phi}_k^{T}\mathbf{M}\boldsymbol{\phi}_k
}.
$$

Finally, the modal damping ratio simplifies to
$$
\zeta_k
=
\frac{1}{2}
\left(
\frac{\alpha}{\omega_k}
+
\beta\,\omega_k
\right).
$$

Since
$$
\lvert \psi_1(l_{\text{beam}}) \rvert \ll \lvert \psi_2(l_{\text{beam}}) \rvert,
$$
the damper is positioned close to a node of the first mode and near an antinode
of the second mode. As a result, the damper primarily influences the second
vibration mode, which corresponds to the higher natural frequency.

Consequently, the damping effect on the second mode is expected to be stronger
than that on the first mode, implying
$$
\zeta_2 > \zeta_1.
$$

Based on this observation, the damping model may be simplified by assuming
pure stiffness-proportional damping for the system, such that
$$
\alpha = 0.
$$

Under this assumption, the modal damping ratio reduces to
$$
\zeta_k = \frac{\beta\,\omega_k}{2}.
$$

Since the maximum excitation frequency available in the experimental setup is
$$
17~\text{Hz},
$$
and the second natural frequency satisfies
$$
\omega_2 > 17~\text{Hz},
$$
it is only possible to experimentally determine the damping ratio of the first
mode at $\omega_1$. The damping ratio of the second mode must therefore be
estimated indirectly.

Using the stiffness-proportional damping relationship, the modal damping ratios
are given by
$$
\zeta_1 = \frac{\beta\,\omega_1}{2},
\qquad
\zeta_2 = \frac{\beta\,\omega_2}{2}.
$$

Eliminating the damping coefficient $\beta$ leads to the following relationship
between the modal damping ratios:
$$
\zeta_2
=
\zeta_1\,\frac{\omega_2}{\omega_1}.
$$

**3.1 The Unforced Repsonse to an Initiail Diplacement of the Spring End of the Beam**



From the modal coordinates of the complete system, the physical coordinates are
related through
$$
\mathbf{z}(t) = \boldsymbol{\Phi}\,\mathbf{q}(t).
$$

This relationship can be written explicitly as
$$
\begin{bmatrix}
z_1(t) \\
z_2(t)
\end{bmatrix}
=
\begin{bmatrix}
\phi_{11} & \phi_{12} \\
\phi_{21} & \phi_{22}
\end{bmatrix}
\begin{bmatrix}
q_1(t) \\
q_2(t)
\end{bmatrix}.
$$

Recall that the truncated two-degree-of-freedom approximation of the beam
displacement is
$$
y(x,t) = z_1(t)\,\psi_1(x) + z_2(t)\,\psi_2(x).
$$

Substituting the modal transformation into this expression allows the mode
shapes of the complete system with attachments to be redefined as
$$
y(x,t)
=
\bigl(\phi_{11}\,q_1(t) + \phi_{12}\,q_2(t)\bigr)\psi_1(x)
+
\bigl(\phi_{21}\,q_1(t) + \phi_{22}\,q_2(t)\bigr)\psi_2(x).
$$

Rearranging terms gives
$$
y(x,t)
=
\underbrace{\bigl(\phi_{11}\,\psi_1(x) + \phi_{21}\,\psi_2(x)\bigr)}_{\psi_1^{*}(x)}
\,q_1(t)
+
\underbrace{\bigl(\phi_{12}\,\psi_1(x) + \phi_{22}\,\psi_2(x)\bigr)}_{\psi_2^{*}(x)}
\,q_2(t).
$$

Let the initial displacement applied at the free end of the beam be
$$
y(l_{\text{beam}},0) = y_0.
$$

Substituting into the above expression yields
$$
y_0
=
\psi_1^{*}(l_{\text{beam}})\,q_1(0)
+
\psi_2^{*}(l_{\text{beam}})\,q_2(0).
$$

Since two unknowns are present and the initial displacement is applied at the
free end, the response may be assumed to be dominated by the first mode only,
such that
$$
q_2(0) = 0.
$$

Under this assumption, the initial modal coordinate becomes
$$
q_1(0)
=
\frac{y_0}{\psi_1^{*}(l_{\text{beam}})}.
$$

In order to determine the modal responses $q_1(t)$ and $q_2(t)$, the unforced
modal equations of motion must be solved. These equations are given by
$$
\mathbf{M}_m\,\ddot{\mathbf{q}}
+
\mathbf{C}_m\,\dot{\mathbf{q}}
+
\mathbf{K}_m\,\mathbf{q}
=
0.
$$

Using the stiffness-proportional damping assumption, the damping matrix may be
expressed as
$$
\mathbf{C}_m = \beta\,\mathbf{K}_m,
$$
which reduces the governing equation to
$$
\mathbf{M}_m\,\ddot{\mathbf{q}}
+
\beta\,\mathbf{K}_m\,\dot{\mathbf{q}}
+
\mathbf{K}_m\,\mathbf{q}
=
0.
$$

The initial conditions associated with the free-vibration response are
$$
\mathbf{q}(0)
=
\begin{bmatrix}
q_1(0) \\
0
\end{bmatrix},
\qquad
\dot{\mathbf{q}}(0)
=
\ddot{\mathbf{q}}(0)
=
\begin{bmatrix}
0 \\
0
\end{bmatrix}.
$$

Here, the proportional damping coefficient is defined as
$$
\beta
=
\frac{2\zeta_1}{\omega_1}
=
\frac{2\zeta_2}{\omega_2}.
$$

Once the modal responses $q_1(t)$ and $q_2(t)$ have been obtained, the final
expression for the free damped vibration response of the system is recovered by
substituting back into
$$
y(x,t)
=
\psi_1^{*}(x)\,q_1(t)
+
\psi_2^{*}(x)\,q_2(t).
$$


```python
y_o = 0.003         # Give a displacement value (m)

phi11 = Phi[0, 0]
phi12 = Phi[0, 1]
phi21 = Phi[1, 0]
phi22 = Phi[1, 1]

# =============== Mode Shape for Complete System ================
def psi_star_1(x):
    return phi11 * psi(x, beta1, l_beam, E*I, k_theta) + \
           phi21 * psi(x, beta2, l_beam, E*I, k_theta)

def psi_star_2(x):
    return phi12 * psi(x, beta1, l_beam, E*I, k_theta) + \
           phi22 * psi(x, beta2, l_beam, E*I, k_theta)

psi_star_1_l_beam = psi_star_1(l_beam)
print(f"psi_star_1(l_beam) = {psi_star_1_l_beam:.4f}")

q1_0 = y_o / psi_star_1_l_beam
print(f"Initial modal coordinate q1_0 = {q1_0:.4f}")

# =========== Diagonalised Matrices M_m, C_m and K_m ============
M_m = Phi.T @ M @ Phi
K_m = Phi.T @ K @ Phi

zeta1 = 0.008
zeta2 = 0.05
beta = 2 * zeta1 / omega_n[0]

C_m = beta * K_m
```

```python
# =============== ODE solution for Free Vibration ================
def system_ode(q_vec, t, M_m_inv, C_m, K_m):
    q1, q2, q1_dot, q2_dot = q_vec
    q = np.array([q1, q2])
    q_dot = np.array([q1_dot, q2_dot])

    # M_m_inv * (-C_m * q_dot - K_m * q)
    q_ddot = M_m_inv @ (-C_m @ q_dot - K_m @ q)

    return [q1_dot, q2_dot, q_ddot[0], q_ddot[1]]

M_m_inv = np.linalg.inv(M_m)

q1_0_val = q1_0
q2_0_val = 0.0

q_dot1_0 = 0.0
q_dot2_0 = 0.0

q_initial = [q1_0_val, q2_0_val, q_dot1_0, q_dot2_0]

t = np.arange(0, 10, 0.001)

solution = odeint(system_ode, q_initial, t, args=(M_m_inv, C_m, K_m))

# Homogeneous solution for Free Vibration
q1h_t = solution[:, 0]
q2h_t = solution[:, 1]
```

```python
# =============== Damped Natural Response at Free End ================
z1_t = np.zeros_like(t)
z2_t = np.zeros_like(t)

for i in range(len(t)):
    q_vector = np.array([q1h_t[i], q2h_t[i]])
    z_vector = Phi @ q_vector
    z1_t[i] = z_vector[0]
    z2_t[i] = z_vector[1]

psi1_l_beam = psi(l_beam, beta1, l_beam, E*I, k_theta)
psi2_l_beam = psi(l_beam, beta2, l_beam, E*I, k_theta)

y_l_beam_t = z1_t * psi1_l_beam + z2_t * psi2_l_beam

plt.figure(figsize=(10, 3))
plt.plot(t, y_l_beam_t)
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.title('Unforced Response at the Free End of the Beam')
plt.grid(True)
plt.show()
```

```python
y_x_t = np.zeros((len(t), len(x)))

# =============== Animated Beam Response y(x,t) ================
for i in range(len(t)):
    def psi_star_1_full(x_vals):
        return (phi11 * psi(x_vals, beta1, l_beam, E*I, k_theta) +
                phi21 * psi(x_vals, beta2, l_beam, E*I, k_theta))

    def psi_star_2_full(x_vals):
        return (phi12 * psi(x_vals, beta1, l_beam, E*I, k_theta) +
                phi22 * psi(x_vals, beta2, l_beam, E*I, k_theta))

    y_x_t[i, :] = (
        psi_star_1_full(x) * q1h_t[i] +
        psi_star_2_full(x) * q2h_t[i]
    )

from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots(figsize=(7, 3))
ax.set_xlim(0, l_beam)

max_disp = np.max(np.abs(y_x_t)) * 1.1
ax.set_ylim(-max_disp, max_disp)

ax.set_xlabel('Position along Beam (m)')
ax.set_ylabel('Displacement (m)')
ax.set_title('Beam Vibration a (Unforced Response)')
ax.grid(True)

line, = ax.plot([], [], lw=2)

def init():
    line.set_data([], [])
    return line,

def animate(i):
    line.set_data(x, y_x_t[i, :])
    return line,

skip = 10
frames_idx = range(0, len(t), skip)

animation = FuncAnimation(
    fig, animate, init_func=init, frames=frames_idx,
    interval=int((t[1] - t[0]) * skip * 1000),
    blit=True)

plt.close(fig)
HTML(animation.to_html5_video()) # Use Colab Notebook
```

**3.2 The Forced Response for Three Choices of the Rotational Speed of the Motor**

Consider the forced vibration response of the complete system, governed by the
modal equations of motion
$$
\mathbf{M}_m\,\ddot{\mathbf{q}}
+
\mathbf{C}_m\,\dot{\mathbf{q}}
+
\mathbf{K}_m\,\mathbf{q}
=
\mathbf{F}_m.
$$

The modal forcing vector is obtained through the modal transformation
$$
\mathbf{F}_m
=
\boldsymbol{\Phi}^{T}\,\mathbf{F}_{\text{exciter}}
$$
where the physical forcing vector applied by the exciter is given by
$$
\mathbf{F}_{\text{exciter}}
=
\begin{bmatrix}
Q(\Omega)\,\psi_1(l_{\text{exciter}})\,\sin(\Omega t) \\
[1mm]
Q(\Omega)\,\psi_2(l_{\text{exciter}})\,\sin(\Omega t)
\end{bmatrix}.
$$

The excitation force amplitude generated by the rotating unbalance of the
exciter motor is expressed as
$$
Q(\Omega) = m_{\text{ecc}}\,r\,\Omega^{2}
$$
where $m_{\text{ecc}}$ denotes the eccentric mass and $r$ represents the offset
radius of the eccentric mass in the servomotor exciter.

The resulting non-homogeneous system of ordinary differential equations can
then be solved to obtain the particular solution
$$
\mathbf{q}_t(t),
$$
corresponding to the forced response.

Once the modal responses $q_1(t)$ and $q_2(t)$ are determined, the displacement
response of the beam is reconstructed using
$$
y(x,t)
=
\psi_1^{*}(x)\,q_1(t)
+
\psi_2^{*}(x)\,q_2(t)
$$
which can be evaluated at any spatial location for response visualization.


```python
# =============== Input Exciter Parameters ================
Omega_1 = 32.472    # Near first resonance
Omega_2 = 100.0     # Intermediate
Omega_3 = 204.048   # Near second resonance

Omega_list = [Omega_1, Omega_2, Omega_3]

m_ecc =  0.0045     # Eccentric mass x 2 (kg)
r     =  0.035      # Offset (m)

def compute_Q(m_ecc, r, Omega):
    #return 1.0
    return 2 * m_ecc * r * Omega**2
```

```python
# =============== ODE solution for Forced Vibration ================
t = np.arange(0, 7.5, 0.001)

def system_ode_forced(q_vec, t, M_m_inv, C_m, K_m, Omega, Q, psi_star_l):
    q1, q2, q1_dot, q2_dot = q_vec
    q = np.array([q1, q2])
    q_dot = np.array([q1_dot, q2_dot])

    f_physical = psi_star_l * Q * np.sin(Omega * t)

    # Forced equation of motion
    q_ddot = M_m_inv @ (f_physical - C_m @ q_dot - K_m @ q)

    return [q1_dot, q2_dot, q_ddot[0], q_ddot[1]]

# Initial conditions
q_initial = [0.0, 0.0, 0.0, 0.0]

psi_star_l = np.array([psi_star_1(l_exciter), psi_star_2(l_exciter)])

#Solutions for Three Motor Speeds
solutions = []

for Omega in Omega_list:

    Q = compute_Q(m_ecc, r, Omega)

    solution = odeint(system_ode_forced, q_initial, t,
    args=(M_m_inv, C_m, K_m, Omega, Q, psi_star_l)
    )

    solutions.append(solution)
```

```python
# =============== Animated Beam Response y(x,t) ================
for idx, Omega in enumerate(Omega_list):

    solution = solutions[idx]
    q1_t = solution[:, 0]
    q2_t = solution[:, 1]

    y_x_t = np.zeros((len(t), len(x)))
    for i in range(len(t)):
        y_x_t[i, :] = (
            psi_star_1_full(x) * q1_t[i] +
            psi_star_2_full(x) * q2_t[i]
        )

    fig, ax = plt.subplots(figsize=(7, 3))
    ax.set_xlim(0, l_beam)

    max_disp = np.max(np.abs(y_x_t)) * 1.1
    ax.set_ylim(-max_disp, max_disp)

    ax.set_xlabel('Position along Beam (m)')
    ax.set_ylabel('Displacement (m)')
    ax.set_title(f'Forced Response, Ω = {Omega:.2f} rad/s')
    ax.grid(True)

    line, = ax.plot([], [], lw=2)

    def init():
        line.set_data([], [])
        return line,

    def animate(i):
        line.set_data(x, y_x_t[i, :])
        return line,

    skip = 10
    frames_idx = range(0, len(t), skip)

    animation = FuncAnimation(
        fig, animate, init_func=init, frames=frames_idx,
        interval=int((t[1] - t[0]) * skip * 1000),
        blit=True
    )

    plt.close(fig)
    display(HTML(animation.to_html5_video())) # Use Colab Notebook
```

**3.3 Theoretical Frequency Response of the System that Captures the First Two Dominant Modes**

The frequency response of the system is defined as the ratio between the
displacement amplitude and the applied force amplitude, given by
$$
G(x,\Omega)
=
\frac{\text{Displacement Amplitude}}{\text{Force Amplitude}}
=
\frac{Y(x,\Omega)}{Q(\Omega)}
$$

For the complete system including attachments, the modal amplitudes and phase
angles are modified from the beam-only expressions to account for the updated
mode shapes and modal masses. The modal amplitude for the $k^{\text{th}}$ mode
is written as
$$
\chi_k^{*}(\Omega)
=
\left|
\frac{
Q(\Omega)\,\psi_k^{*}(l_{\text{exciter}})
}{
M_{m_k}
\sqrt{
(\omega_k^2 - \Omega^2)^2
+
4\zeta_k^2 \Omega^2 \omega_k^2
}
}
\right|
$$

The corresponding phase angle is given by
$$
\phi_k^{*}(\Omega)
=
\arctan
\left(
\frac{2\zeta_k \Omega \omega_k}{\omega_k^2 - \Omega^2}
\right).
$$

Here, $\omega_k$ and $\zeta_k$ denote the natural frequencies and damping ratios
of the complete system with attachments.

The displacement response of the beam may then be expressed as
$$
y(x,t)
=
\psi_1^{*}(x)\,q_1(t)
+
\psi_2^{*}(x)\,q_2(t)
$$
which, under harmonic excitation, becomes
$$
y(x,t)
=
\chi_1^{*}(\Omega)\cos\!\bigl(\Omega t + \phi_1^{*}(\Omega)\bigr)\,\psi_1^{*}(x)
+
\chi_2^{*}(\Omega)\cos\!\bigl(\Omega t + \phi_2^{*}(\Omega)\bigr)\,\psi_2^{*}(x)
$$

Using the complex exponential representation
$$
\cos\!\bigl(\Omega t + \phi_k^{*}(\Omega)\bigr)
=
\Re\!\left\{
e^{j(\Omega t + \phi_k^{*}(\Omega))}
\right\}
$$
the displacement can be rewritten as
$$
y(x,t)
=
\Re\!\left\{
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j(\Omega t + \phi_1^{*}(\Omega))}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j(\Omega t + \phi_2^{*}(\Omega))}
\right\}
$$

Factoring out the time-dependent term yields
$$
y(x,t)
=
\Re\!\left\{
\left[
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j\phi_1^{*}(\Omega)}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j\phi_2^{*}(\Omega)}
\right]
e^{j\Omega t}
\right\}
$$

The spatial frequency-domain displacement is therefore defined as
$$
Y(x,\Omega)
=
\chi_1^{*}(\Omega)\,\psi_1^{*}(x)\,e^{j\phi_1^{*}(\Omega)}
+
\chi_2^{*}(\Omega)\,\psi_2^{*}(x)\,e^{j\phi_2^{*}(\Omega)}
$$

Finally, the frequency response function of the system is obtained as
$$
G(x,\Omega)
=
\frac{Y(x,\Omega)}{Q(\Omega)}
$$


Recall that the original definition of the modified modal amplitude
$$
\chi_k^{*}(\Omega)
$$
contains only magnitude information, since
$$
\left|
(\omega_k^2 - \Omega^2) + j(2\zeta_k \Omega \omega_k)
\right|
=
\sqrt{(\omega_k^2 - \Omega^2)^2 + 4\zeta_k^2 \Omega^2 \omega_k^2}.
$$

Similarly, the modified phase term
$$
\phi_k^{*}(\Omega)
$$
represents the phase angle, defined as
$$
\phi_k^{*}(\Omega)
=
\arg\!\left(
(\omega_k^2 - \Omega^2) + j(2\zeta_k \Omega \omega_k)
\right).
$$

However, when constructing the frequency response function, both magnitude and
phase information must be preserved. This requires reverting to the full
complex representation of the frequency response.

Accordingly, the frequency-domain displacement can be written as
$$
Y(x,\Omega)
=
\frac{
Q(\Omega)\,\psi_1^{*}(l_{\text{exciter}})
}{
M_{m_1}
\bigl[(\omega_1^2 - \Omega^2) + j(2\zeta_1 \Omega \omega_1)\bigr]
}
\,\psi_1^{*}(x)
+
\frac{
Q(\Omega)\,\psi_2^{*}(l_{\text{exciter}})
}{
M_{m_2}
\bigl[(\omega_2^2 - \Omega^2) + j(2\zeta_2 \Omega \omega_2)\bigr]
}
\,\psi_2^{*}(x).
$$

Dividing through by the excitation force amplitude $Q(\Omega)$ yields
$$
\frac{Y(x,\Omega)}{Q(\Omega)}
=
\frac{1}{Q(\Omega)}
\left[
\frac{
Q(\Omega)\,\psi_1^{*}(l_{\text{exciter}})\,\psi_1^{*}(x)
}{
M_{m_1}
\bigl[(\omega_1^2 - \Omega^2) + j(2\zeta_1 \Omega \omega_1)\bigr]
}
+
\frac{
Q(\Omega)\,\psi_2^{*}(l_{\text{exciter}})\,\psi_2^{*}(x)
}{
M_{m_2}
\bigl[(\omega_2^2 - \Omega^2) + j(2\zeta_2 \Omega \omega_2)\bigr]
}
\right].
$$

The frequency response function of the system is therefore defined as
$$
G(x,\Omega)
=
\frac{
\psi_1^{*}(l_{\text{exciter}})\,\psi_1^{*}(x)
}{
M_{m_1}
\bigl[(\omega_1^2 - \Omega^2) + j(2\zeta_1 \Omega \omega_1)\bigr]
}
+
\frac{
\psi_2^{*}(l_{\text{exciter}})\,\psi_2^{*}(x)
}{
M_{m_2}
\bigl[(\omega_2^2 - \Omega^2) + j(2\zeta_2 \Omega \omega_2)\bigr]
}.
$$

Using this expression, the magnitude response is obtained from
$$
|G(x,\Omega)|,
$$
while the phase response is given by
$$
\arg\!\bigl(G(x,\Omega)\bigr).
$$


```python
# ======================= Input Beam Positions for FRF =========================
x1 = l_beam
x2 = l_damper
x3 = 0.495       # Near node of second mode

x_vals_f = np.array([x1, x2, x3])

psi1_x = np.array([psi_star_1(x) for x in x_vals_f])
psi2_x = np.array([psi_star_2(x) for x in x_vals_f])

psi1_le = psi_star_1(l_exciter)
psi2_le = psi_star_2(l_exciter)

omega_1, omega_2 = omega_n[0], omega_n[1]

M_m1 = M_m[0, 0]
M_m2 = M_m[1, 1]

# =========================== Evaluation of FRF ================================
def G(x_idx, Omega):

    denom1 = (omega_1**2 - Omega**2) + 1j * (2 * zeta1 * Omega * omega_1)
    denom2 = (omega_2**2 - Omega**2) + 1j * (2 * zeta2 * Omega * omega_2)

    term1 = (psi1_le * psi1_x[x_idx]) / (M_m1 * denom1)
    term2 = (psi2_le * psi2_x[x_idx]) / (M_m2 * denom2)

    return term1 + term2
```

```python
# ============ Frequency Response at 3 Selected Positions on Beam ==============
Omega_vals = np.linspace(0.01, 300, 600)

G_all = np.zeros((3, len(Omega_vals)), dtype=complex)

for i in range(3):
    for j, Omega in enumerate(Omega_vals):
        G_all[i, j] = G(i, Omega)

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Magnitude in dB
for i in range(3):
    axs[0].plot(Omega_vals, 20 * np.log10(np.abs(G_all[i])), label=f"$x_{i+1}$")
axs[0].set_xlabel(r"$\Omega$")
axs[0].set_ylabel(r"$20\log_{10}|G(x,\Omega)|$ [dB]")
axs[0].set_title("FRF Magnitude (dB)")
axs[0].legend()
axs[0].grid(True)

# Phase (unwrap for clarity)
for i in range(3):
    axs[1].plot(Omega_vals, np.unwrap(np.angle(G_all[i])), label=f"$x_{i+1}$")
axs[1].set_xlabel(r"$\Omega$")
axs[1].set_ylabel(r"$\arg(G(x,\Omega))$ [rad]")
axs[1].set_title("FRF Phase")
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

Based on the results obtained above, a noticeable discrepancy is observed
between the theoretical predictions and the experimental measurements. This
difference primarily arises from deviations between the assumed effective
stiffness of the system and its actual physical value.

According to the TecQuipment User Guide~[1], a more accurate theoretical estimate
of the first natural frequency is obtained using \emph{Rigid Beam Theory}
compared to the Euler--Bernoulli beam theory employed in this report. This
indicates that, in this particular case, the more advanced Euler--Bernoulli
model provides a less accurate prediction of the system dynamics. In other
words, the Euler--Bernoulli formulation does not fully represent the dominant
physical behaviour of the system.

The natural frequency of the system follows the relationship
$$
\omega_n \propto \sqrt{\frac{k}{m}},
$$
and the discrepancy arises because flexural deformation of the beam has been
explicitly included in the Euler--Bernoulli formulation. In practice, however,
the flexural compliance of the beam is negligible in comparison to the
compliance of the attached spring at low frequencies. Consequently, the rigid
beam assumption yields a more realistic representation of the physical system.

As a result, the inclusion of beam flexure in the theoretical model leads to an
underestimation of the fundamental natural frequency and a corresponding
overestimation of the maximum steady-state displacement amplitude. This occurs
because the model accounts for additional flexural motion that does not
significantly contribute to the actual system response.

An additional source of error arises from the idealisation of the base boundary
condition associated with the rotational stiffness. During rotation, the beam
may experience deformation such that the actual spring displacement deviates
from the assumed relationship
$$
y_{\text{spring}} = \theta\,l_{\text{spring}}.
$$
Furthermore, the lumped mass effect of the exciter motor tends to suppress
flexural deformation, causing the fundamental mode to behave more like a rigid
body motion rather than a bending-dominated response.

Considering these factors, a more appropriate modelling approach for this
experiment would be the use of the \emph{Rayleigh--Ritz method}. In this hybrid
formulation, the fundamental frequency is obtained using rigid body theory,
while higher-order modes are estimated using Euler--Bernoulli beam theory. The
combined results yield a more accurate representation of the system dynamics,
albeit at the expense of increased mathematical complexity.
