## 2-DOF Model Derivation



**Define coordinates:**  
Primary beam displacement $y_1$ (at mid-span), absorber displacement $y_2$ (relative to beam).

### Equations of Motion (Undamped, Small Angle)

**Primary beam:**  
$$ m_1 \ddot{y_1} + k_\text{beam} y_1 + k_\text{abs} (y_1 - y_2) = Q(t) \sin \omega t $$  
$$ \implies (m_1 \ddot{y_1} + 2 k_\text{abs} y_1 - 2 k_\text{abs} y_2) + (k_\text{beam} - k_\text{abs}) y_1 = Q \sin \omega t $$

**Absorber:**  
$$ m_2 \ddot{y_2} + k_\text{abs} (y_2 - y_1) = 0 $$  
$$ \implies 2 m_2 \ddot{y_2} + 2 k_\text{abs} (y_2 - y_1) = 0 $$

Where:  
- $m_1 = m_\text{exciter} + 0.8 m_\text{beam} + m_\text{abs} \approx 5.368$ kg (Rayleigh effective)  
- $m_2 = 2 m_\text{mass} + 0.47 m_\text{abs,beam} \approx 0.142$ kg (tunable)  
- $k_\text{beam} = 4.74 \times 10^4$ N/m, $k_\text{abs} = 1.09 \times 10^3$ N/m

### Matrix Form (Harmonic Solution: $y_1 = A \sin \omega t$, $y_2 = B \sin \omega t$)

$$
\begin{bmatrix}
k_\text{beam} + 2k_\text{abs} - m_1 \omega^2 & -2k_\text{abs} \\
-2k_\text{abs} & 2k_\text{abs} - 2m_2 \omega^2
\end{bmatrix}
\begin{bmatrix}
A \\
B
\end{bmatrix}
=
\begin{bmatrix}
Q \\
0
\end{bmatrix}
$$

### Natural Frequencies (Free Vibration, $Q=0$)

Determinant $= 0$:  
$$ (2m_1 m_2) \omega^4 - [2m_1 k_\text{abs} + 2m_2 (k_\text{beam} + 2k_\text{abs})] \omega^2 + 2 k_\text{beam} k_\text{abs} = 0 $$

Quadratic solution:  
$$ \omega_{1,2}^2 = \frac{b \pm \sqrt{b^2 - 4ac}}{2a} $$  
$a = 2m_1 m_2$, $b = 2m_1 k_\text{abs} + 2m_2 (k_\text{beam}+2k_\text{abs})$, $c = 2 k_\text{beam} k_\text{abs}$  

**Modes:** Lower $\omega_1 \approx 13$ Hz, upper $\omega_2 \approx 16.5$ Hz (matches Exp. 6).

### Antiresonance ($A=0$)

$$ \omega = \sqrt{\frac{k_\text{abs}}{m_2}} \approx \sqrt{\frac{1.2 \times 10^3}{0.142}} \approx 14.8 \text{ rad/s} \ (f \approx 14 \text{ Hz}) $$

Absorber tunes to primary mode: $B = -Q / (2 k_\text{abs})$.

This captures beam flexure + absorber flexure as the first two dominant modes for small amplitudes.

## Damping Ratio Estimation

For the TM1016V simply supported beam + vibration absorber (2-DOF system), estimate damping ratios $\zeta_1$ and $\zeta_2$ for the first two dominant modes using logarithmic decrement from free vibration response.

### Method 1: Logarithmic Decrement (Free Vibration)

From underdamped free vibration: $x(t) = A e^{-\zeta \omega_n t} \cos(\omega_d t + \phi)$

**Logarithmic decrement:**
$$ \delta = \ln\left(\frac{x_n}{x_{n+1}}\right) = \frac{2\pi \zeta}{\sqrt{1-\zeta^2}} \approx 2\pi \zeta \quad (\text{for small } \zeta) $$

**Damping ratio:**
$$ \zeta \approx \frac{\delta}{2\pi} = \frac{1}{2\pi} \ln\left(\frac{x_n}{x_{n+1}}\right) $$

### Method 2: 2-DOF Modal Damping

For coupled 2-DOF system, measure free vibration at **two locations**:
- $y_1(t)$: Primary beam displacement (VDAS displacement sensor)
- $y_2(t)$: Absorber relative displacement (if measurable, or derive from beam motion)

**Peak-picking from each mode:**
1. **Mode 1** ($\omega_1 \approx 13$ Hz): In-phase motion ($y_1, y_2 > 0$)
2. **Mode 2** ($\omega_2 \approx 16.5$ Hz): Out-of-phase motion ($y_1, y_2 < 0$)

For each mode $i$:
$$ \zeta_i = \frac{1}{2\pi} \ln\left(\frac{A_{i,n}}{A_{i,n+1}}\right) $$

### Expected Values (from TM1016V typical results)

|------|----------------|-------------------|

### 1. Unforced Response (Free Vibration - Initial Displacement)

**Setup:** Simply supported beam + vibration absorber, displace spring end, release from rest.

**Plot:** Time history of $y_1(t)$ (primary beam displacement at exciter location)


```python
# Week 1 - Activity 3 Plot 1: FIXED VERSION
# Add these imports FIRST (run this cell before plots)

import numpy as np
import matplotlib.pyplot as plt

# Create sample data for free vibration (2-DOF beating)
t_free = np.linspace(0, 2, 1000)
omega1, omega2 = 13*2*np.pi, 16.5*2*np.pi  # rad/s
zeta1, zeta2 = 0.005, 0.008

# 2-DOF free vibration response (beats)
y1_free = 5 * np.exp(-zeta1*omega1*t_free) * np.sin(omega1*t_free) + \
          3 * np.exp(-zeta2*omega2*t_free) * np.sin(omega2*t_free + np.pi/2)

plt.figure(figsize=(10,6))
plt.plot(t_free, y1_free, 'b-', linewidth=2, label='$y_1(t)$ - Beam')
plt.xlabel('Time [s]')
plt.ylabel('Displacement [mm]')
plt.title('Free Vibration Response (Initial Displacement)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
