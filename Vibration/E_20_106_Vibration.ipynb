## 2-DOF Vibration Model Assumptions

- The beam is rigid and pinned at the left end
- Small-angle vibration: sin(θ) ≈ θ
- A linear spring supports the right end of the beam
- The motor is modeled as a lumped mass mounted at a distance x m from the pin
- Motor mounting flexibility is modeled as a vertical spring km
- Damping is neglected
- The unbalanced motor provides a harmonic excitation force


part B

```python
import sympy as sp
t = sp.symbols('t')

theta = sp.Function('theta')(t)
y = sp.Function('y')(t)

Ib, mb, mm = sp.symbols('Ib mb mm')
L, xm = sp.symbols('L xm')
ks, km = sp.symbols('ks km')

Fu = sp.Function('Fu')(t)
```
