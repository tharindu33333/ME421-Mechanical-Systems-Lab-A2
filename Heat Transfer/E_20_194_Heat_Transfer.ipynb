# Cooling Load Analysis for Milk Storage

A factory involved with the manufacturing of dairy products is required to store milk received from collecting centres in a cool room until it is delivered to the factory on the next day. Containers made of aluminium are received at around 1400 hours each day.

The average temperature of collected milk is in the range of $35^\circ C$ -- $40^\circ C$. It is required to cool the bulk milk to $15^\circ C$ within two hours and maintain it at the same temperature until the delivery time of 1400 hours on the next day.

**Given Data**

- Number of milk containers collected per day = 40  
- Weight of an empty container = $2 \, \text{kg}$  
- Weight of milk in a container = $5 \, \text{kg}$  
- Specific heat capacity of container material = $921.1 \, \text{J/kg·K}$  
- Specific heat capacity of milk = $4000 \, \text{J/kg·K}$  

**Required Tasks**

1. Propose a method to determine the feasibility of the available unit.
2. Experiment with the cooling unit in the Heat Engine Laboratory and obtain necessary data.
3. Determine the developments required if storage increases to 80 milk containers per day.
4. Prepare a report based on the findings.


## Cooling Load Calculation for Milk Storage

### 1. System Parameters

Number of containers: $n = 40$  
Mass of milk per container: $m_m = 5 \text{ kg}$  
Mass of empty container: $m_c = 2 \text{ kg}$  
Specific heat of milk: $c_m = 4000 \text{ J/kgK}$  
Specific heat of aluminum: $c_c = 921.1 \text{ J/kgK}$  
Target temperature: $T_f = 15^\circ C$  
Cooling time: $t = 2 \text{ hours} = 7200 \text{ s}$

---

### 2. Case A — Average Initial Temperature

$$
T_{avg} = \frac{35 + 40}{2} = 37.5^\circ C
$$


Heat removed:

$$
Q = n \Big[(m_m c_m + m_c c_c)(T_{avg} - T_f)\Big]
$$

$$
Q = 40 \Big[(5 \times 4000 + 2 \times 921.1)(37.5 - 15)\Big]
$$

$$
Q = 19{,}657{,}980 \text{ J} \approx 19.66 \text{ MJ}
$$

Cooling load:

$$
\dot{Q} = \frac{Q}{t} = \frac{19{,}657{,}980}{7200}
$$

$$
\dot{Q} \approx 2.73 \text{ kW}
$$

---

### 3. Case B — Maximum Initial Temperature

Initial temperature: $T_i = 40^\circ C$

Heat removed:

$$
Q = 40 \Big[(5 \times 4000 + 2 \times 921.1)(40 - 15)\Big]
$$

$$
Q = 21{,}842{,}200 \text{ J} \approx 21.84 \text{ MJ}
$$

Cooling load:

$$
\dot{Q} = \frac{Q}{t} = \frac{21{,}842{,}200}{7200}
$$

$$
\dot{Q} \approx 3.03 \text{ kW}
$$


# THEORY  
## Experimental Method for Evaluating Cooling Unit Performance

This section describes the experimental method used to estimate the actual cooling capacity $\dot{Q}_{act}$ of the laboratory air-conditioning unit and to judge whether it is suitable for the required milk cooling application.

---

## 1. Required Measurements

To evaluate the system performance, the following geometric and thermodynamic quantities must be measured.

**Duct Dimensions:** Measure the width $(w)$ and height $(h)$ of the inlet air duct and compute the cross-sectional area using

$$
A = w \times h
$$

**Air Velocity $(v)$:** Measure inlet air velocity using an anemometer. Take several readings at different points and use the average value to reduce flow variation error.

**Air State Conditions:** Record Dry Bulb Temperature $(T_{db})$ and Wet Bulb Temperature $(T_{wb})$ at two locations:

- Inlet state — air entering the evaporator  
- Outlet state — air leaving the evaporator  

---

## 2. Psychrometric Background

Cooling capacity depends on the thermodynamic condition of moist air. Since atmospheric air contains both dry air and water vapor, its total energy (enthalpy) includes both sensible and latent heat components.

### Vapor Pressure and Humidity Ratio

First determine the saturation vapor pressure at wet bulb temperature using the Magnus–Tetens relation:

$$
P_{swb} = 0.61078 \exp\left(\frac{17.27\,T_{wb}}{T_{wb}+237.3}\right)
$$

Then estimate the actual vapor pressure:

$$
P_v = P_{swb} - 0.000662\, P_{atm}\,(T_{db}-T_{wb})
$$

Humidity ratio is calculated as

$$
\omega = \frac{0.62194\,P_v}{P_{atm} - P_v}
$$

---

## 3. Enthalpy of Moist Air

The specific enthalpy of moist air is computed from

$$
h = c_{pa} T_{db} + \omega \left(h_{fg} + c_{pv} T_{db}\right)
$$

where

- $c_{pa} \approx 1.006$ kJ/kg·K — specific heat of dry air  
- $h_{fg} \approx 2501$ kJ/kg — latent heat of vaporization at $0^\circ C$  
- $c_{pv} \approx 1.86$ kJ/kg·K — specific heat of water vapor

---

## 4. Performance Calculations

### Air Mass Flow Rate

The air mass flow rate through the duct is estimated by

$$
\dot{m}_a = \rho \, A \, v
$$

with air density approximately

$$
\rho \approx 1.2 \ \text{kg/m}^3
$$

---

### Actual Cooling Capacity

The real cooling capacity of the unit is obtained from the enthalpy drop across the evaporator:

$$
\dot{Q}_{act} = \dot{m}_a (h_{in} - h_{out})
$$

---

---

## 6. Suitability Decision Rule

The cooling unit is considered acceptable if its measured capacity satisfies the required peak load:

$$
\text{Required Load} = 3.03 \ \text{kW}
$$

Decision condition:

- If $\dot{Q}_{act} \ge 3.03$ kW — Unit is adequate  
- If $\dot{Q}_{act} < 3.03$ kW — Unit needs improvement

## Experimental Calculations and Thermodynamic Basis

This section explains the mathematical and thermodynamic relationships used to estimate the cooling capacity of the laboratory air-conditioning unit using the measured experimental data.

---

### Air Mass Flow Rate ($\dot{m}_a$)

The air mass flow rate is a key parameter in determining the rate of heat transfer. It is computed from the measured inlet air velocity and the duct cross-sectional area.

$$
\dot{m}_a = \rho \, A \, V_{in}
$$

where:

- $\rho$ — air density (approximately $1.2 \ \text{kg/m}^3$)  
- $A$ — duct cross-sectional area ($\text{m}^2$)  
- $V_{in}$ — measured inlet air velocity (m/s)

---

### Moist Air Enthalpy Calculation ($h$)

To evaluate the total heat removed from the air (including both sensible and latent components), the dry bulb temperature ($T_{db}$) and wet bulb temperature ($T_{wb}$) measurements are used.

---

#### Step A — Saturation Vapor Pressure at Wet Bulb Temperature

Using the Magnus–Tetens correlation:

$$
P_{swb} = 0.61078 \, \exp\!\left(\frac{17.27 \, T_{wb}}{T_{wb} + 237.3}\right)
$$

---

#### Step B — Actual Vapor Pressure

Correcting for the difference between dry bulb and wet bulb temperatures:

$$
P_v = P_{swb} - 0.000662 \, P_{atm} \, (T_{db} - T_{wb})
$$

where:

- $P_{atm} = 101.325 \ \text{kPa}$ — atmospheric pressure

---

#### Step C — Humidity Ratio

The humidity ratio (mass of water vapor per mass of dry air) is calculated by

$$
\omega = 0.62194 \, \frac{P_v}{P_{atm} - P_v}
$$

---

#### Step D — Specific Enthalpy of Moist Air

The specific enthalpy of moist air is then determined from

$$
h = (c_{p,a} \, T_{db}) + \omega \, (h_{fg} + c_{p,v} \, T_{db})
$$

where:

- $c_{p,a} = 1.006 \ \text{kJ/kg·K}$ — specific heat of dry air  
- $h_{fg} = 2501 \ \text{kJ/kg}$ — latent heat of vaporization  
- $c_{p,v} = 1.86 \ \text{kJ/kg·K}$ — specific heat of water vapor

---

### Cooling Capacity (Heat Removal Rate) ($\dot{Q}$)

The experimental cooling capacity of the unit is obtained from the enthalpy difference between the inlet (return air) and outlet (evaporator exit) conditions:

$$
\dot{Q} = \dot{m}_a \, (h_{in} - h_{out})
$$

where:

- $h_{in}$ — enthalpy calculated using $T_1$ (return air DB) and $T_2$ (return air WB)  
- $h_{out}$ — enthalpy calculated using $T_5$ (exit air DB) and $T_6$ (exit air WB)

---


```python
import numpy as np
import pandas as pd

# 1. Experimental Data
data = {
    "fan_speed": [3, 4, 5, 6, 7, 8, 9, 10],
    "T1_db_in": [29.9, 30.0, 29.2, 29.2, 29.4, 29.0, 28.7, 29.4],  # Return Air DB
    "T2_wb_in": [25.3, 24.9, 23.7, 23.3, 23.2, 23.1, 22.9, 23.2],  # Return Air WB
    "T5_db_out": [24.3, 20.4, 20.8, 21.4, 21.5, 21.2, 21.3, 19.9],  # Exit DB
    "T6_wb_out": [19.7, 16.6, 16.2, 18.0, 18.2, 17.6, 17.4, 16.5],  # Exit WB
    "V_in": [0.6, 2.8, 5.6, 8.8, 9.3, 9.6, 9.8, 9.5]  # Air velocity (m/s)
}

df = pd.DataFrame(data)

# 2. Thermodynamic Constants
P_atm = 101.325  # kPa
rho_air = 1.2    # kg/m3
D_1 = 0.11       # duct diameter (m)
A_duct = np.pi*(D_1**2)/4  # duct area (m2)

# 3. Enthalpy Calculation
def get_enthalpy(T_db, T_wb):
    """Calculates specific enthalpy (kJ/kg)."""
    P_swb = 0.61078 * np.exp((17.27 * T_wb) / (T_wb + 237.3))  # Saturation vapor pressure
    P_v = P_swb - 0.000662 * P_atm * (T_db - T_wb)             # Actual vapor pressure
    w = 0.62194 * P_v / (P_atm - P_v)                          # Humidity ratio
    h = 1.006 * T_db + w * (2501 + 1.86 * T_db)                # Enthalpy in kJ/kg
    return h

# 4. Calculations
df['h_in'] = df.apply(lambda row: get_enthalpy(row['T1_db_in'], row['T2_wb_in']), axis=1)
df['h_out'] = df.apply(lambda row: get_enthalpy(row['T5_db_out'], row['T6_wb_out']), axis=1)
df['m_dot'] = rho_air * A_duct * df['V_in']  # Mass flow rate (kg/s)
df['Q_dot_kW'] = df['m_dot'] * (df['h_in'] - df['h_out'])  # Cooling load (kW)

# 5. Display Final Results
print(df[['fan_speed', 'V_in', 'm_dot', 'Q_dot_kW']])
```

- **Operational Setting:** While air velocity slightly peaked at fan speed 9, the lowest exit temperatures ($T_5 = 19.9^\circ\text{C}$, $T_6 = 16.5^\circ\text{C}$) occurred at speed 10, indicating the most effective cooling.

### Thermodynamic Analysis: Instantaneous Cooling Rate

The heat removal rate at each time step is calculated as follows:

#### Step 1 — Constant Mass Flow Rate

Using the optimized inlet velocity from Experiment 01:

$$
\dot{m}_a = \rho \cdot A \cdot V_{in}
$$

where:

- $\rho \approx 1.2 \ \text{kg/m}^3$  
- $A \approx 0.0095 \ \text{m}^2$ (circular duct area)  
- $V_{in} = 9.5 \ \text{m/s}$

---

#### Step 2 — Instantaneous Specific Enthalpy

For each time step $t$, calculate the specific enthalpy at the evaporator inlet ($T_3$, $T_4$) and outlet ($T_5$, $T_6$) using the standard psychrometric relations discussed previously.

---

#### Step 3 — Instantaneous Heat Removal Rate

The cooling capacity at each time interval is:

$$
\dot{Q}_t = \dot{m}_a \cdot (h_{in,t} - h_{out,t})
$$

where:

- $h_{in,t}$ — enthalpy of inlet air at time $t$  
- $h_{out,t}$ — enthalpy of outlet air at time $t$

This sequence allows tracking the unit's pull-down performance and assessing its ability to reach the target storage temperature.
