Part 1: Thermal Load Analysis

1.1 Problem Description

The objective is to determine the cooling load required to cool milk containers received at 1400hrs. The bulk must be cooled from the initial temperature range of $35^{\circ}\text{C}-40^{\circ}\text{C}$ to a target temperature of $15^{\circ}\text{C}$ within a 2-hour window.

1.2 Given ParametersThe following data from the factory specifications are used for the calculation:

Number of containers ($n$): $40$

Mass of milk per container ($m_{milk\_unit}$): $5\text{ kg}$

Mass of empty container ($m_{cont\_unit}$): $2\text{ kg}$

Specific Heat of Milk ($C_{p,m}$): $4000\text{ J/kg K}$

Specific Heat of Aluminum ($C_{p,c}$): $921.1\text{ J/kg K}$ 1.3

Thermal Load Calculation

We assume the worst-case scenario for the initial temperature ($T_1 = 40^{\circ}\text{C}$) to find the maximum required cooling capacity.

Total Mass Calculations:$$M_{milk} = 40 \times 5\text{ kg} = 200\text{ kg}$$$$M_{container} = 40 \times 2\text{ kg} = 80\text{ kg}$$

Temperature Difference ($\Delta T$):$$\Delta T = T_{initial} - T_{final} = 40^{\circ}\text{C} - 15^{\circ}\text{C} = 25\text{ K}$$

Total Heat Removal ($Q_{total}$):$$Q_{total} = (M_{milk} \cdot C_{p,m} \cdot \Delta T) + (M_{container} \cdot C_{p,c} \cdot \Delta T)$$$$Q_{total} = (200 \cdot 4000 \cdot 25) + (80 \cdot 921.1 \cdot 25)$$$$Q_{total} = 20,000,000\text{ J} + 1,842,200\text{ J} = 21,842,200\text{ J}$$

Required Cooling Power ($P_{req}$):
Since the cooling must be completed in $2\text{ hours}$ ($7200\text{ seconds}$):
$$P_{req} = \frac{Q_{total}}{t} = \frac{21,842,200\text{ J}}{7200\text{ s}}$$$$P_{req} \approx \mathbf{3.033 \text{ kW}}$$
