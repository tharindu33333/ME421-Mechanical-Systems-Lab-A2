## 1. Cooling Load Calculation

To determine the suitability of the unit, we calculate the total heat energy ($Q_{total}$) that must be removed to cool the milk and containers within **2 hours**.

---

### 1.1 Parameters & Data
| Parameter | Symbol | Value |
| :--- | :--- | :--- |
| Number of containers | $n$ | 40 |
| Total mass of milk | $m_m$ | $200 \text{ kg}$ |
| Total mass of containers | $m_c$ | $80 \text{ kg}$ |
| Specific heat of milk | $C_{p,m}$ | $4000 \text{ J/kg K}$ |
| Specific heat of Aluminum | $C_{p,c}$ | $921.1 \text{ J/kg K}$ |
| Temperature Change | $\Delta T$ | $22.5 \text{ K}$ |

---

### 1.2 Energy Equations
The total heat removed is the sum of the heat lost by the milk ($Q_{milk}$) and the containers ($Q_{cont}$):

$$Q_{total} = (m_m \cdot C_{p,m} \cdot \Delta T) + (m_c \cdot C_{p,c} \cdot \Delta T)$$

**Step-by-step Calculation:**
* **Milk:** $200 \times 4000 \times 22.5 = 18,000,000 \text{ J}$
* **Containers:** $80 \times 921.1 \times 22.5 = 1,657,980 \text{ J}$
* **Total ($Q_{total}$):** $19,657,980 \text{ J} \approx \mathbf{19.66 \text{ MJ}}$

---

### 1.3 Required Cooling Power ($\dot{Q}$)
The unit must achieve this cooling within $t = 7200 \text{ s}$ (2 hours):

$$\dot{Q}_{req} = \frac{Q_{total}}{t} = \frac{19,657,980}{7200}$$

$$\dot{Q}_{req} \approx 2730.28 \text{ W} = \mathbf{2.73 \text{ kW}}$$

> **Conclusion:** The unit requires a minimum sensible cooling capacity of **2.73 kW**, excluding external heat gains.

---

### 1.4 Worst-Case Scenario ($T_i = 40^\circ\text{C}$)
For an initial temperature of $40^\circ\text{C}$, the temperature change becomes $\Delta T = 25 \text{ K}$.

**Total Heat Removal ($Q$):**
$$Q = 40 \times [(5 \cdot 4000 + 2 \cdot 921.1) \times 25]$$
$$Q = 21,842,200 \text{ J} \approx \mathbf{21.84 \text{ MJ}}$$

**Required Cooling Load ($\dot{Q}_{max}$):**
$$\dot{Q}_{max} = \frac{21,842,200}{7200} \approx \mathbf{3.03 \text{ kW}}$$

### 2.2 Mathematical Framework
The cooling capacity is determined by the mass flow rate of the air and the change in its specific enthalpy.

**I. Mass Flow Rate ($\dot{m}_a$):**
$$\dot{m}_a = \rho \cdot A \cdot \bar{v}$$
> *Where $\rho$ is the air density ($\approx 1.2 \text{ kg/m}^3$) and $\bar{v}$ is the mean velocity.*

**II. Energy Balance ($\dot{Q}_{act}$):**
The total heat extraction (sensible + latent) is given by:
$$\dot{Q}_{act} = \dot{m}_a \Delta h = \dot{m}_a (h_{1} - h_{2})$$
> *Enthalpy values ($h$) are derived from psychrometric data based on measured $T_{db}$ and $T_{wb}$.*

## 5. Analytical Framework: Thermodynamic Evaluation

This section establishes the mathematical procedure for converting raw experimental sensor data into the system's **Actual Cooling Capacity** ($\dot{Q}_{act}$).

---

### 5.1 Mass Flow Analysis ($\dot{m}_a$)
The mass flow rate is the fundamental driver of the total heat exchange. It is calculated using the continuity equation based on the intake duct geometry:

$$\dot{m}_a = \rho \cdot A \cdot V_{in}$$




### 5.2 Psychrometric Modeling for Specific Enthalpy ($h$)
To account for both **sensible** (temperature) and **latent** (moisture) heat removal, we derive the specific enthalpy ($h$) using the Magnus-Tetens approximation.

#### **I. Vapor Pressure Derivation**
First, we determine the saturation vapor pressure ($P_{swb}$) at the wet-bulb temperature, followed by the actual vapor pressure ($P_v$):

$$P_{swb} = 0.61078 \cdot \exp\left(\frac{17.27 \cdot T_{wb}}{T_{wb} + 237.3}\right)$$

$$P_v = P_{swb} - 0.000662 \cdot P_{atm} \cdot (T_{db} - T_{wb})$$

#### **II. Humidity Ratio ($\omega$) and Enthalpy ($h$)**
The humidity ratio defines the mass of water vapor per unit mass of dry air, which allows for the calculation of total energy content:

$$\omega = 0.62194 \cdot \frac{P_v}{P_{atm} - P_v}$$

$$h = (c_{p,a} \cdot T_{db}) + \omega \cdot (h_{fg} + c_{p,v} \cdot T_{db})$$

> **Constants:** $c_{p,a} = 1.006 \text{ kJ/kg}\cdot\text{K}$, $h_{fg} = 2501 \text{ kJ/kg}$, $c_{p,v} = 1.86 \text{ kJ/kg}\cdot\text{K}$, and $P_{atm} = 101.325 \text{ kPa}$.

---

### 5.3 System Capacity Quantization ($\dot{Q}_{act}$)
The net cooling effect is the product of the air mass flow and the enthalpy gradient across the evaporator coils ($h_{inlet}$ to $h_{outlet}$):

$$\dot{Q}_{act} = \dot{m}_a \cdot (h_{inlet} - h_{outlet})$$



---

### 5.4 Performance Metric
The success of each fan speed setting is measured by the **Performance Ratio** ($\eta$):

$$\eta = \frac{\dot{Q}_{act}}{\dot{Q}_{req}}$$

*where $\dot{Q}_{req} = 3.03 \text{ kW}$.*

```python
import numpy as np
import pandas as pd

def calculate_enthalpy(Tdb, Twb):
    """
    Calculates specific enthalpy (h) in kJ/kg based on Dry Bulb and Wet Bulb temperatures.
    Uses the formulas established in Section 5.
    """
    Patm = 101.325  # kPa

    # I. Saturation Vapor Pressure at Wet Bulb (Pswb)
    Pswb = 0.61078 * np.exp((17.27 * Twb) / (Twb + 237.3))

    # II. Actual Vapor Pressure (Pv)
    Pv = Pswb - 0.000662 * Patm * (Tdb - Twb)

    # III. Humidity Ratio (w)
    w = 0.62194 * (Pv / (Patm - Pv))

    # IV. Total Specific Enthalpy (h)
    # cp_a = 1.006, hfg = 2501, cp_v = 1.86
    h = (1.006 * Tdb) + w * (2501 + 1.86 * Tdb)

    return h

# Experimental Data from your spreadsheet
data = {
    'Speed': range(1, 11),
    'T3_db': [29.1, 29.1, 29.5, 29.6, 29.3, 29.3, 29.3, 29.1, 28.9, 29.3],
    'T4_wb': [21.2, 21.1, 21.5, 21.8, 21.2, 21.1, 21.1, 21.0, 20.9, 21.1],
    'T5_db': [26.4, 26.7, 24.3, 20.4, 20.8, 21.4, 21.5, 21.2, 21.3, 19.9],
    'T6_wb': [19.9, 20.0, 19.7, 16.6, 16.2, 18.0, 18.2, 17.6, 17.4, 16.5],
    'Vin':   [0.0, 0.0, 0.6, 2.8, 5.6, 8.8, 9.3, 9.0, 9.8, 9.5]
}

df = pd.DataFrame(data)

# Constants for Capacity Calculation
rho = 1.2  # kg/m3
A = 0.05   # EDIT THIS: Enter your actual Duct Area in m^2

# Process Calculations
df['h_in'] = df.apply(lambda row: calculate_enthalpy(row['T3_db'], row['T4_wb']), axis=1)
df['h_out'] = df.apply(lambda row: calculate_enthalpy(row['T5_db'], row['T6_wb']), axis=1)
df['m_dot'] = rho * A * df['Vin']
df['Q_act_kW'] = df['m_dot'] * (df['h_in'] - df['h_out'])

# Comparison against Design Load (3.03 kW)
df['Requirement_Met'] = df['Q_act_kW'] >= 3.03

# Displaying Results
print("--- Evaporator Performance Analysis ---")
display(df[['Speed', 'Vin', 'h_in', 'h_out', 'Q_act_kW', 'Requirement_Met']])
```

## 6. Scaling Analysis: Developments for 80 Containers

This section outlines the necessary upgrades required to maintain the 2-hour cooling mandate if the daily throughput is doubled to **80 containers**.

---

### 6.1 Revised Thermal Load Requirement
Doubling the product mass directly doubles the total energy ($Q$) that must be removed within the same $t = 7200 \text{ s}$ window.

**New Total Heat Energy ($Q_{new}$):**
$$Q_{new} = 2 \cdot Q_{max} = 2 \cdot 21,842,200 \text{ J}$$
$$Q_{new} = 43,684,400 \text{ J} \approx \mathbf{43.68 \text{ MJ}}$$

**Revised Cooling Power ($\dot{Q}_{req, 80}$):**
$$\dot{Q}_{req, 80} = \frac{Q_{new}}{7200} \approx \mathbf{6.06 \text{ kW}}$$

---

### 6.2 Hardware & Engineering Developments

#### **I. Unit Capacity Upgrade**
The current unit's theoretical requirement was $3.03 \text{ kW}$. To handle $6.06 \text{ kW}$ plus environmental heat gains, the following is required:
* **Compressor Scaling:** Upgrade to a **2.5 Ton (approx. 9 kW)** commercial unit.
* **Safety Factor:** A larger buffer is needed to handle the increased latent heat from opening the door twice as often for 80 units.

#### **II. Airflow Management**
To ensure all 80 containers reach $15^\circ\text{C}$ simultaneously, the mass flow rate ($\dot{m}_a$) must increase to prevent thermal stratification.
* **Requirement:** $\dot{m}_{a, new} \approx 2 \cdot \dot{m}_{a, current}$
* **Development:** Installation of a **ducted plenum system** to distribute chilled air evenly across the larger stack, avoiding "dead zones" in the center of the storage area.
