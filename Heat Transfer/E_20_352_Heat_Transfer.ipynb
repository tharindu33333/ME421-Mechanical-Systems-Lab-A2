```python
# ======================================================
# Cooling Load Calculation (Rewritten – Same Results)
# ======================================================

# -------------------------
# GIVEN DATA
# -------------------------
N = 40                     # number of containers
m_c = 2                    # mass of empty container (kg)
m_m = 5                    # mass of milk per container (kg)

cp_c = 921.1               # specific heat of container (J/kgK)
cp_m = 4000.0              # specific heat of milk (J/kgK)

T_target = 15.0            # storage temperature (°C)
cooling_time_hr = 2        # cooling duration (hours)

# Convert time to seconds
time_sec = cooling_time_hr * 3600


# -------------------------
# FUNCTION: Cooling Load
# -------------------------
def cooling_load(T_initial):

    # Temperature difference
    delta_T = T_initial - T_target

    # Total thermal capacity (m·cp) of all containers + milk
    thermal_capacity = N * (m_m * cp_m + m_c * cp_c)

    # Total heat removed (J)
    Q_total = thermal_capacity * delta_T

    # Required power (W)
    power_watts = Q_total / time_sec

    # Convert to kW
    return power_watts / 1000


# -------------------------
# TEMPERATURE SCENARIOS
# -------------------------
T_avg = (35 + 40) / 2
T_max = 40.0

load_avg = cooling_load(T_avg)
load_max = cooling_load(T_max)


# -------------------------
# OUTPUT
# -------------------------
print("════════════════════════════════════════")
print(" Cooling Load Calculation Summary")
print("════════════════════════════════════════")

print(f"Scenario 1: Average Initial Temperature ({T_avg} °C)")
print(f"  Required Cooling Load : {load_avg:10.3f} kW")

print(f"\nScenario 2: Maximum Initial Temperature ({T_max} °C)")
print(f"  Required Cooling Load : {load_max:10.3f} kW")

print("════════════════════════════════════════")
```

2. Psychrometric Principles
The effective cooling capacity relies on the moist air properties. As air comprises dry air and water vapor, its total energy content (enthalpy) combines sensible heat (from temperature variation) and latent heat (from moisture phase transition).

Vapor Pressure and Humidity Ratio: To calculate specific enthalpy ( h ), first compute the humidity ratio ( ω ), indicating water vapor mass per unit mass of dry air. Use the saturation vapor pressure ( Pswb ) at wet bulb temperature with the Magnus-Tetens formula:
Pswb=0.61078exp(17.27⋅TwbTwb+237.3)
The true vapor pressure ( Pv ) accounts for the wet-dry bulb depression:
Pv=Pswb−0.000662⋅Patm⋅(Tdb−Twb)
Then, obtain the humidity ratio ( ω ):
ω=0.62194⋅PvPatm−Pv

Specific Enthalpy Formula: Total enthalpy ( h ) in kJ/kg follows this energy equation:
h=cpaTdb+ω(hfg+cpvTdb)
Where:

cpa≈1.006 kJ/kg⋅K  (Specific heat of dry air)
hfg≈2501 kJ/kg  (Enthalpy of vaporization at  0∘C )
cpv≈1.86 kJ/kg⋅K  (Specific heat of water vapor)

Actual Cooling Capacity ( Q˙act ): Heat extraction rate equals mass flow times enthalpy reduction across the evaporator:
Q˙act=m˙a×(hin−hout)

## Test Evaluation and Thermodynamic Framework

### Air Mass Flow Rate ($\dot{m}_a$)
Mass flow rate serves as the core variable for heat transfer computations. It comes from the intake air velocity measurement:

$$\dot{m}_a = \rho \cdot A \cdot V_{in}$$

* **$\rho$**: Air density ($\approx 1.2 \text{ kg/m}^3$).
* **$A$**: Duct cross-sectional area ($m^2$).
* **$V_{in}$**: Recorded air intake velocity ($m/s$).

---

### Specific Enthalpy Computation ($h$)
To quantify total energy extraction (sensible plus latent heat) from the air, utilize the Dry Bulb ($T_{db}$) and Wet Bulb ($T_{wb}$) readings.

#### **Step A: Saturation Vapor Pressure ($P_{swb}$)**
Apply the Magnus-Tetens formula at Wet Bulb temperature:
$$P_{swb} = 0.61078 \cdot \exp\left(\frac{17.27 \cdot T_{wb}}{T_{wb} + 237.3}\right)$$

#### **Step B: True Vapor Pressure ($P_v$)**
Use the psychrometric equation to include evaporative cooling:
$$P_v = P_{swb} - 0.000662 \cdot P_{atm} \cdot (T_{db} - T_{wb})$$
* **$P_{atm}$**: Atmospheric pressure ($101.325 \text{ kPa}$).

#### **Step C: Humidity Ratio ($\omega$)**
$$\omega = 0.62194 \cdot \frac{P_v}{P_{atm}-P_v}$$

#### **Step D: Specific Enthalpy ($h$)**
Total enthalpy per kg of dry air in $\text{kJ/kg}$:
$$h = (c_{p,a} \cdot T_{db}) + \omega \cdot (h_{fg} + c_{p,v} \cdot T_{db})$$

* **$c_{p,a}$**: $1.006 \text{ kJ/kg}\cdot\text{K}$ (Dry air specific heat).
* **$h_{fg}$**: $2501 \text{ kJ/kg}$ (Vaporization latent heat).
* **$c_{p,v}$**: $1.86 \text{ kJ/kg}\cdot\text{K}$ (Water vapor specific heat).



---

### Heat Extraction Rate Calculation ($\dot{Q}$)
Experimental cooling capacity reflects the enthalpy drop from **Return Air** (Inlet) to **Evaporator Exit** (Outlet):

$$\dot{Q} = \dot{m}_a \cdot (h_{in} - h_{out})$$

* **$h_{in}$**: From $T_1$ (Return DB) and $T_2$ (Return WB).
* **$h_{out}$**: From $T_5$ (Exit DB) and $T_6$ (Exit WB).

---

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ======================================================
# 1. Experimental Data
# ======================================================
data = {
    "fan_speed": [3, 4, 5, 6, 7, 8, 9, 10],
    "T1_db_in": [29.9, 30.0, 29.2, 29.2, 29.4, 29.0, 28.7, 29.4],  # Return Air DB
    "T2_wb_in": [25.3, 24.9, 23.7, 23.3, 23.2, 23.1, 22.9, 23.2],  # Return Air WB
    "T5_db_out": [24.3, 20.4, 20.8, 21.4, 21.5, 21.2, 21.3, 19.9],  # Exit DB
    "T6_wb_out": [19.7, 16.6, 16.2, 18.0, 18.2, 17.6, 17.4, 16.5],  # Exit WB
    "V_in": [0.6, 2.8, 5.6, 8.8, 9.3, 9.6, 9.8, 9.5]                # Air velocity (m/s)
}

df = pd.DataFrame(data)

# ======================================================
# 2. Thermodynamic Constants
# ======================================================
P_atm = 101.325   # Atmospheric pressure (kPa)
rho_air = 1.2     # Air density (kg/m³)
D_1 = 0.11        # Duct diameter (m)

# Duct cross-sectional area
A_duct = np.pi * (D_1 ** 2) / 4

# ======================================================
# 3. Psychrometric Enthalpy Function
# ======================================================
def get_enthalpy(T_db, T_wb):
    """
    Calculates specific enthalpy (kJ/kg)
    using psychrometric relations.
    """

    # Saturation vapor pressure at wet-bulb temperature (kPa)
    P_swb = 0.61078 * np.exp((17.27 * T_wb) / (T_wb + 237.3))

    # Actual vapor pressure (kPa)
    P_v = P_swb - 0.000662 * P_atm * (T_db - T_wb)

    # Humidity ratio
    w = 0.62194 * P_v / (P_atm - P_v)

    # Moist air enthalpy (kJ/kg)
    h = 1.006 * T_db + w * (2501 + 1.86 * T_db)

    return h

# ======================================================
# 4. Calculations
# ======================================================
# Enthalpy at inlet and outlet
df['h_in'] = df.apply(lambda row: get_enthalpy(row['T1_db_in'], row['T2_wb_in']), axis=1)
df['h_out'] = df.apply(lambda row: get_enthalpy(row['T5_db_out'], row['T6_wb_out']), axis=1)

# Mass flow rate (kg/s)
df['m_dot'] = rho_air * A_duct * df['V_in']

# Cooling capacity / Heat removal rate (kW)
df['Q_dot_kW'] = df['m_dot'] * (df['h_in'] - df['h_out'])

# ======================================================
# 5. Visualization
# ======================================================
plt.figure(figsize=(10, 6))

plt.plot(
    df['fan_speed'],
    df['Q_dot_kW'],
    marker='s',
    linestyle='-',
    linewidth=2,
    markersize=7,
    color='purple'   # Graph color
)

plt.title(r'Heat Removal Rate vs. Fan Speed (Inlet: $T_1, T_2$)', fontsize=14)
plt.xlabel('Fan Speed Setting', fontsize=12)
plt.ylabel('Heat Removal Rate (kW)', fontsize=12)

# Target cooling capacity line
plt.axhline(y=3.03, linestyle='--', color='red', label='Target Max (3.03 kW)')

plt.grid(True, alpha=0.3)
plt.legend(['Measured Cooling Capacity'], loc='upper left')

plt.show()

# ======================================================
# 6. Printed Results
# ======================================================
print("════════════════════════════════════════")
print(" Experimental Heat Removal Rates        ")
print("════════════════════════════════════════")

for _, row in df[['fan_speed', 'V_in', 'Q_dot_kW']].iterrows():
    print(f"Fan Speed {int(row['fan_speed']):<2}: "
          f"V_in={row['V_in']:<4.1f} m/s, "
          f"Q_dot={row['Q_dot_kW']:>7.3f} kW")

print("════════════════════════════════════════")
```

### Computational Method for Time-Varying Cooling Rate

To compute cooling capacity per interval, follow this thermodynamic process:

**Step 1: Compute Fixed Mass Flow Rate ($\dot{m}_a$)**
Apply optimal velocity from Experiment 01 ($V_{in} = 9.5$ m/s) and duct area ($A \approx 0.0095$ m$^2$):
$$\dot{m}_a = \rho \cdot A \cdot V_{in}$$
*Where $\rho \approx 1.2$ kg/m$^3$.*

**Step 2: Find Time-Specific Enthalpy ($h_{t}$)**
At each timestep ($t$), derive enthalpy for inlet ($T_3, T_4$) and outlet ($T_5, T_6$) via prior psychrometric method.


**Step 3: Determine Time-Varying Heat Extraction ($\dot{Q}_t$)**
Instantaneous cooling power equals mass flow times evaporator enthalpy difference:
$$\dot{Q}_t = \dot{m}_a \cdot (h_{in,t} - h_{out,t})$$

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 1. Experimental Data (Experiment 02)
# =====================================================
time = np.arange(1, 16)

T3 = np.array([27.4,27.0,26.7,26.4,26.1,25.9,25.7,25.5,25.4,25.2,25.1,24.9,24.8,24.6,24.5])
T4 = np.array([19.9,19.6,19.3,18.9,18.7,18.5,18.3,18.2,18.0,17.9,17.7,17.6,17.5,17.4,17.3])
T5 = np.array([20.3,20.2,20.0,19.8,19.7,19.5,19.3,19.2,18.9,18.8,18.7,18.6,18.4,18.4,18.2])
T6 = np.array([16.4,16.4,16.2,16.1,16.0,15.8,15.6,15.5,15.4,15.3,15.2,15.0,15.0,14.9,14.8])

# =====================================================
# 2. System Constants
# =====================================================
P_atm = 101.325        # kPa
rho_air = 1.2          # kg/m^3
duct_diameter = 0.11   # m
velocity = 9.5         # m/s (optimum fan speed)

area = np.pi * duct_diameter**2 / 4
m_dot = rho_air * area * velocity   # mass flow rate (kg/s)

# =====================================================
# 3. Psychrometric Enthalpy Function
# =====================================================
def enthalpy_moist_air(T_db, T_wb):

    p_sat_wb = 0.61078 * np.exp((17.27 * T_wb) / (T_wb + 237.3))
    p_vapor = p_sat_wb - 0.000662 * P_atm * (T_db - T_wb)

    humidity_ratio = 0.62194 * p_vapor / (P_atm - p_vapor)

    h = 1.006 * T_db + humidity_ratio * (2501 + 1.86 * T_db)
    return h

# =====================================================
# 4. Calculations
# =====================================================
h_in = np.array([enthalpy_moist_air(db, wb) for db, wb in zip(T3, T4)])
h_out = np.array([enthalpy_moist_air(db, wb) for db, wb in zip(T5, T6)])

# Heat removal rate (kW)
Q_dot = m_dot * (h_in - h_out)

# Cooling rate (°C/min)
cooling_rate = -np.gradient(T3, time)

# =====================================================
# 5. Plotting (COLORS CHANGED)
# =====================================================
fig, ax_left = plt.subplots(figsize=(10, 6))

# ---- Heat Removal Rate (Primary Axis)
ax_left.set_xlabel('Time (minutes)', fontsize=12)
ax_left.set_ylabel(r'Heat Removal Rate ($\dot{Q}$) (kW)',
                   fontsize=12, color='royalblue')

line1, = ax_left.plot(
    time, Q_dot,
    marker='o',
    linewidth=2,
    color='royalblue',
    label=r'Heat Removal Rate ($\dot{Q}$)'
)

ax_left.tick_params(axis='y', labelcolor='royalblue')
ax_left.grid(True, linestyle='--', alpha=0.7)

# ---- Cooling Rate (Secondary Axis)
ax_right = ax_left.twinx()
ax_right.set_ylabel(r'Cooling Rate ($dT/dt$) ($^\circ$C/min)',
                    fontsize=12, color='darkorange')

line2, = ax_right.plot(
    time, cooling_rate,
    marker='s',
    linewidth=2,
    color='darkorange',
    label=r'Cooling Rate ($dT/dt$)'
)

ax_right.tick_params(axis='y', labelcolor='darkorange')

# ---- Title & Legend
plt.title('Dynamic Cooling Performance Over Time (Experiment 02)', fontsize=14)

lines = [line1, line2]
labels = [l.get_label() for l in lines]
ax_left.legend(lines, labels, loc='upper right')

plt.tight_layout()
plt.savefig('cooling_performance_plot.png')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# ======================================================
# 1. Experimental Data (Experiment 02)
# ======================================================
time_data = np.arange(1, 16)

T3_data = np.array([
    27.4, 27.0, 26.7, 26.4, 26.1,
    25.9, 25.7, 25.5, 25.4, 25.2,
    25.1, 24.9, 24.8, 24.6, 24.5
])

Q_data = np.array([
    1.169, 1.111, 1.077, 1.054, 1.011,
    0.985, 0.963, 0.925, 0.884, 0.852,
    0.821, 0.795, 0.771, 0.765, 0.762
])

# ======================================================
# 2. Experimental Thermal Capacity (C_exp)
# ======================================================
Q_avg = Q_data.mean()

cooling_rate = (T3_data[0] - T3_data[-1]) / (time_data[-1] - time_data[0])

C_exp = (Q_avg * 60) / cooling_rate   # kJ/K

# ======================================================
# 3. Linear Model  Q = m*T + c
# ======================================================
reg_model = LinearRegression()
reg_model.fit(T3_data.reshape(-1, 1), Q_data)

m = reg_model.coef_[0]
c = reg_model.intercept_

# ======================================================
# 4. Simulation Function (120 minutes)
# ======================================================
def simulate_system(T_start, duration=120):

    Q_history = []
    T_history = [T_start]

    T_current = T_start

    for _ in range(duration):
        Q_inst = m * T_current + c
        Q_history.append(Q_inst)

        # temperature update
        T_current -= (Q_inst * 60) / C_exp
        T_history.append(T_current)

    return np.array(T_history), np.array(Q_history)

# ======================================================
# 5. Visualization (Colors Changed)
# ======================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12))

# ---------- Heat Removal Rate ----------
ax1.plot(
    time_data, Q_data,
    marker='o',
    linestyle='-',
    color='navy',
    label='Actual Experiment (27.4°C Start)'
)

simulation_times_Q = np.arange(1, 121)
start_temperatures = [35.0, 37.5, 40.0]
prediction_colors = ['teal', 'goldenrod', 'crimson']

for col, T_start in zip(prediction_colors, start_temperatures):
    _, Q_sim = simulate_system(T_start)
    ax1.plot(simulation_times_Q, Q_sim,
             linestyle='--',
             color=col,
             label=f'Predicted: {T_start}°C Start')

ax1.axhline(3.03, linestyle='-', color='black',
            label='Target Cooling Load (3.03 kW)')

ax1.set_title(r'Simulated Heat Removal Rate ($\dot{Q}$) Over 120 Minutes', fontsize=16)
ax1.set_xlabel('Time (minutes)')
ax1.set_ylabel(r'Heat Removal Rate $\dot{Q}$ (kW)')
ax1.set_xlim(0, 120)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend(loc='upper right', fontsize=10)

# ---------- Temperature Pull-Down ----------
ax2.plot(
    time_data, T3_data,
    marker='D',
    linestyle='-',
    color='navy',
    label='Actual Experiment (27.4°C Start)'
)

simulation_times_T = np.arange(0, 121)

for col, T_start in zip(prediction_colors, start_temperatures):
    T_sim, _ = simulate_system(T_start)
    ax2.plot(simulation_times_T, T_sim,
             linestyle='--',
             color=col,
             label=f'Predicted: {T_start}°C Start')

ax2.axhline(15, linestyle='-', color='black',
            label='Target Temperature (15°C)')

ax2.set_title(r'Simulated Room Temperature ($T_3$) Pull-Down Over 120 Minutes', fontsize=16)
ax2.set_xlabel('Time (minutes)')
ax2.set_ylabel(r'Temperature $T_3$ (°C)')
ax2.set_xlim(0, 120)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.legend(loc='upper right', fontsize=10)

plt.tight_layout()
plt.show()
```

Therefore it can be confidently claimed that the air conditioning unit in the laboratory is not capable of cooling the storage within the specified period of time of 2hrs under any circumstances.

### Scaling for 80 Containers Daily: Required Upgrades

Doubling to 80 containers raises thermal mass to ~$1747.4 \text{ kJ/K}$. For $25^\circ\text{C}$ drop in 2 hours:  
$$\dot{Q}_{80} = \frac{1747.4 \times 25}{7200} \approx 6.06 \text{ kW}$$

**Current Gap**: Lab unit max ~$1.24{-}1.68 \text{ kW}$ (simulated), shortfall of $4.5 \text{ kW}$.

**Proposed Modifications**:
  - **Refrigeration Boost**: Add parallel circuit or upgrade compressor/evaporator to ≥$7 \text{ kW}$ (15% margin for walls).
  - **Flow Enhancement**: High-velocity ducting + blower to eliminate dead zones in expanded load.
  - **TES Integration**: Ice/chilled-water bank for off-peak "coolth" storage, rapid discharge in 2-hr window.
  - **Leak Prevention**: Strip curtains/air-lock to cut infiltration during prolonged 80-container loading.
